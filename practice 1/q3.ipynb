```python
# PROJECT 3: Zomato Delivery Time Estimator

try:
    # User Inputs
    distance = float(input("Enter distance in km: "))
    traffic_level = int(input("Enter traffic level (1=Low, 2=Medium, 3=High): "))

    # Match Case for Traffic Level
    match traffic_level:
        case 1:
            speed = 40
            traffic = "Low"
        case 2:
            speed = 25
            traffic = "Medium"
        case 3:
            speed = 15
            traffic = "High"
        case _:
            raise ValueError("Invalid traffic level")

    # Calculate Delivery Time
    delivery_time = (distance / speed) * 60

    # Add Preparation Time
    total_time = delivery_time + 15

    # Display Result
    print("\n===== Delivery Details =====")
    print(f"Distance       : {distance:.2f} km")
    print(f"Traffic Level  : {traffic}")
    print(f"Estimated Delivery Time: {total_time:.0f} minutes")

except ValueError:
    print("Error: Please enter a valid traffic level (1, 2, or 3).")
```
