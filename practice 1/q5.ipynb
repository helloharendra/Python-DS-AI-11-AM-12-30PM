```python
deposit(account, 2000)
deposit(account, 1500)
deposit(account, 3000)

# 2 Withdrawals
withdraw(account, 1000)
withdraw(account, 2500)

# Balance
print()
check_balance(account)

# History
transaction_history(account)
```
