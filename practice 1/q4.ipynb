```python
# PROJECT 4: Number Pattern Analyzer

print("Call/WhatsApp:: +91 7991327022 | Email: support@nayaprayog.com | Website: https://nayaprayog.com/academy")

print("\nNaya Prayog Academy")
print("From Code to Career Empowering Digital Futures!")
print("Website: https://nayaprayog.com/academy")

numbers = []

print("\nEnter 20 Numbers:")

# Collect 20 Numbers with Exception Handling
while len(numbers) < 20:
    try:
        num = int(input(f"Enter Number {len(numbers) + 1}: "))
        numbers.append(num)
    except ValueError:
        print("Invalid input! Please enter a valid number.")

# List Comprehensions
even_numbers = [num for num in numbers if num % 2 == 0]
odd_numbers = [num for num in numbers if num % 2 != 0]
squared_values = [num ** 2 for num in numbers]

# Average Calculation
total = sum(numbers)
average = total / len(numbers)

above_average = [num for num in numbers if num > average]

# Find Maximum Without max()
maximum = numbers[0]
for num in numbers:
    if num > maximum:
        maximum = num

# Find Minimum Without min()
minimum = numbers[0]
for num in numbers:
    if num < minimum:
        minimum = num

# Prime Number Check
prime_numbers = []

for num in numbers:
    if num > 1:
        is_prime = True

        for i in range(2, int(num ** 0.5) + 1):
            if num % i == 0:
                is_prime = False
                break

        if is_prime:
            prime_numbers.append(num)

# Results
print("\n" + "=" * 50)
print("NUMBER PATTERN ANALYZER REPORT")
print("=" * 50)

print(f"\nOriginal Numbers : {numbers}")
print(f"Even Numbers     : {even_numbers}")
print(f"Odd Numbers      : {odd_numbers}")
print(f"Squared Values   : {squared_values}")
print(f"Above Average    : {above_average}")

print("\nStatistics")
print("-" * 20)
print(f"Maximum Number : {maximum}")
print(f"Minimum Number : {minimum}")
print(f"Average        : {average:.2f}")

print("\nPrime Numbers")
print("-" * 20)

if prime_numbers:
    print(prime_numbers)
else:
    print("No prime numbers found.")
```
