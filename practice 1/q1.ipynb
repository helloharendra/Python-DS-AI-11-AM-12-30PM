```python
# PROJECT 1: Student Grade Manager

print("Call/WhatsApp:: +91 7991327022 | Email: support@nayaprayog.com | Website: https://nayaprayog.com/academy")
print("\nNaya Prayog Academy")
print("From Code to Career Empowering Digital Futures!")
print("Website: https://nayaprayog.com/academy")

try:
    # Student Details
    name = input("\nEnter Student Name: ")

    # Subject Marks
    math = float(input("Enter Math Marks: "))
    science = float(input("Enter Science Marks: "))
    english = float(input("Enter English Marks: "))
    hindi = float(input("Enter Hindi Marks: "))
    computer = float(input("Enter Computer Marks: "))

    # Total and Percentage
    total = math + science + english + hindi + computer
    percentage = (total / 500) * 100

    # Grade Calculation
    if percentage >= 90:
        grade = "A"
    elif percentage >= 75:
        grade = "B"
    elif percentage >= 60:
        grade = "C"
    elif percentage >= 40:
        grade = "D"
    else:
        grade = "F (Fail)"

    # Report Card
    print("\n")
    print("Student Report Card")
    print("-------------------")
    print(f"Name       : {name}")
    print(f"Total      : {total:.0f} / 500")
    print(f"Percentage : {percentage:.2f}%")
    print(f"Grade      : {grade}")

except ValueError:
    print("\nError: Please enter valid numeric marks only.")
```
