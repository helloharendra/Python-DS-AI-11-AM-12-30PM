```python
# PROJECT 2: E-Commerce Cart System

# Product Catalog
product_catalog = {
    "Laptop": 50000,
    "Mobile": 20000,
    "Headphones": 1500,
    "Keyboard": 1200,
    "Mouse": 800,
    "Monitor": 10000
}

# Empty Cart
cart = []

while True:
    print("\n===== E-Commerce Cart System =====")
    print("1. View Products")
    print("2. Add Product to Cart")
    print("3. Remove Product from Cart")
    print("4. View Cart & Total Bill")
    print("5. Exit")

    choice = input("Enter your choice (1-5): ")

    if choice == "1":
        print("\nAvailable Products:")
        for product, price in product_catalog.items():
            print(f"{product} : ₹{price}")

    elif choice == "2":
        product_name = input("Enter product name to add: ")

        if product_name in product_catalog:
            cart.append(product_name)
            print(f"{product_name} added to cart.")
        else:
            print("Product not found in catalog.")

    elif choice == "3":
        product_name = input("Enter product name to remove: ")

        if product_name in cart:
            cart.remove(product_name)
            print(f"{product_name} removed from cart.")
        else:
            print(f"{product_name} is not in the cart.")

    elif choice == "4":
        if len(cart) == 0:
            print("Cart is empty.")
        else:
            print("\nCurrent Cart:")
            total = 0

            for item in cart:
                price = product_catalog[item]
                total += price
                print(f"{item} : ₹{price}")

            gst = total * 0.18
            final_bill = total + gst

            print(f"\nSubtotal : ₹{total:.2f}")
            print(f"GST (18%) : ₹{gst:.2f}")
            print(f"Total Bill : ₹{final_bill:.2f}")

    elif choice == "5":
        print("Thank you for shopping!")
        break

    else:
        print("Invalid choice. Please select between 1 and 5.")
```
