# Nested Loops

```python
for i in range(5):
    for j in range(4):
        for k in range(3):
            print(i,j,k)
```

# Loop Control Statements

### Break
- Stops loop immediately

```python
for i in range(500):
    if i==8:
        break
    print(i)
```

### continue
- Skips current iteration

```python
for i in range(50):
    if i>49:
        continue
    print(i)
```

```python
for i in range(15):
    if i<=10:
        
        print('continue works')
        continue
    break
```

### pass
- Placeholder statement

```python
for i in range(5):
    print(i)
    pass
```
