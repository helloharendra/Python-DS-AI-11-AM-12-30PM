# 10. Pivot Tables & Cross-Tabs

### Pivot tables and cross-tabs are used to summarize data and analyze relationships between columns.

```python
sales = pd.DataFrame({
    'Date': ['2024-01-01', '2024-01-01', '2024-01-02', '2024-01-02'],
    'Product': ['A', 'B', 'A', 'B'],
    'Sales': [100, 150, 200, 130],
    'Region': ['East', 'West', 'East', 'West']
})
```

What this does

Creates a DataFrame named sales.

Output
Date	Product	Sales	Region
2024-01-01	A	100	East
2024-01-01	B	150	West
2024-01-02	A	200	East
2024-01-02	B	130	West

```python
pivot = pd.pivot_table(
    sales,
    values='Sales',
    index='Product',
    columns='Region',
    aggfunc='sum',
    fill_value=0
)
```

Meaning of Each Parameter
Parameter	Meaning
sales	Source DataFrame
values='Sales'	Use Sales column values
index='Product'	Rows become Product
columns='Region'	Columns become Region
aggfunc='sum'	Sum sales values
fill_value=0	Replace missing values with 0
Result
Product	East	West
A	300	0
B	0	280
Explanation
Product A sold:
100 + 200 = 300 in East
Product B sold:
150 + 130 = 280 in West

This is similar to Excel Pivot Tables.

```python
# Create sample sales data
sales = pd.DataFrame({
    'Date': ['2024-01-01', '2024-01-01', '2024-01-02', '2024-01-02'],
    'Product': ['A', 'B', 'A', 'B'],
    'Sales': [100, 150, 200, 130],
    'Region': ['East', 'West', 'East', 'West']
})

pivot = pd.pivot_table(sales, values='Sales', index='Product', columns='Region', aggfunc='sum', fill_value=0)
print("Pivot Table (Sales by Product and Region):\n", pivot)

# Cross-tab (frequency table)
crosstab = pd.crosstab(sales['Product'], sales['Region'])
print("\nCross-tab (counts):\n", crosstab)
```

# Purpose

Counts frequency of combinations.

Output
Product	East	West
A	2	0
B	0	2
Explanation
Product A appears 2 times in East
Product B appears 2 times in West

Unlike pivot table:

Pivot Table → summarizes values
Crosstab → counts occurrences

# 11. Apply & Lambda Functions

Used for custom operations on columns.

```python
# ------------------------------
# 11. APPLY & LAMBDA FUNCTIONS
# ------------------------------
print("\n" + "="*60)
print("11. APPLY FOR CUSTOM OPERATIONS")
print("="*60)

def grade(score):
    if score >= 90:
        return 'A'
    elif score >= 80:
        return 'B'
    else:
        return 'C'

df['Grade'] = df['Score'].apply(grade)
print("Grades added:\n", df[['Name', 'Score', 'Grade']])
```

```python
# Apply to multiple columns at once
df['Sum_Cols'] = df[['Age', 'Score']].apply(lambda row: row['Age'] + row['Score'], axis=1)
print("\nSum of Age and Score:\n", df[['Age', 'Score', 'Sum_Cols']])
```
