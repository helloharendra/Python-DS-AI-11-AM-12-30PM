```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

print('Seaborn version:', sns.__version__)

# Set global style
sns.set_theme(style='whitegrid', palette='muted')
plt.rcParams['figure.figsize'] = (9, 5)
```

```python
# Create a comprehensive sample dataset
np.random.seed(42)
n = 200
df = pd.DataFrame({
    'Name': [f'Emp_{i}' for i in range(n)],
    'Department': np.random.choice(['IT', 'HR', 'Finance', 'Marketing', 'Operations'], n),
    'Gender': np.random.choice(['Male', 'Female'], n),
    'Age': np.random.randint(22, 55, n),
    'Experience': np.random.randint(1, 20, n),
    'Salary': np.random.randint(30000, 150000, n),
    'Performance': np.random.choice(['Excellent', 'Good', 'Average', 'Poor'], n, p=[0.2, 0.4, 0.3, 0.1]),
    'City': np.random.choice(['Delhi', 'Mumbai', 'Bangalore', 'Chennai', 'Hyderabad'], n)
})

# Add a calculated column
df['Salary'] = df['Salary'] + df['Experience'] * 2000

print('Dataset Shape:', df.shape)
print(df.head())
```

---
## 1. Available Themes

```python
# Show 4 different themes
themes = ['darkgrid', 'whitegrid', 'dark', 'white']
fig, axes = plt.subplots(1, 4, figsize=(16, 3))

for ax, theme in zip(axes, themes):
    with sns.axes_style(theme):
        data = np.random.normal(0, 1, 100)
        ax.hist(data, bins=10, color='green', edgecolor='red')
        ax.set_title('demo')

plt.suptitle('Seaborn Themes', fontsize=14)
plt.tight_layout()
plt.show()
```

---
## 2. Distribution Plots

Understanding how data is spread.

```python
# histplot - Distribution of salaries
plt.figure(figsize=(20, 10))
sns.histplot(
    data=df,
    x='Salary',
    bins=30,
    kde=True,
    color='steelblue',
    edgecolor='white'
)
plt.axvline(df['Salary'].mean(), color='red', linestyle='--', label=f'Mean: Rs{df["Salary"].mean():,.0f}')
plt.title('Salary Distribution', fontsize=16)
plt.xlabel('Salary (Rs)')
plt.ylabel('Count')
plt.tight_layout()
plt.show()
```

```python
# KDE Plot - Smooth density curve
plt.figure(figsize=(10, 5))
sns.kdeplot(
    data=df,
    x='Salary',
    hue='Department',
    fill=True,
    alpha=0.4,
    linewidth=1.5
)
plt.title('Salary Distribution by Department (KDE)', fontsize=16)
plt.xlabel('Salary (Rs)')
plt.tight_layout()
plt.show()
```

```python
# Box Plot
plt.figure(figsize=(11, 5))
sns.boxplot(
    data=df,
    x='Department',
    y='Salary',
    palette='Set2',
    width=0.5
)
plt.title('Salary Distribution by Department (Box Plot)', fontsize=16)
plt.xlabel('Department')
plt.ylabel('Salary (Rs)')
plt.xticks(rotation=15)
plt.tight_layout()
plt.show()
print('Box Plot shows: Median (line), Q1-Q3 (box), Whiskers, Outliers (dots)')
```

```python
# Violin Plot = Box Plot + KDE
plt.figure(figsize=(11, 5))
sns.violinplot(
    data=df,
    x='Department',
    y='Salary',
    palette='pastel',
    inner='box'
)
plt.title('Salary Distribution by Department (Violin Plot)', fontsize=16)
plt.xlabel('Department')
plt.ylabel('Salary (Rs)')
plt.tight_layout()
plt.show()
```

---
## 3. Categorical Plots

```python
# Bar Plot (shows mean by default)
plt.figure(figsize=(10, 5))
sns.barplot(
    data=df,
    x='Department',
    y='Salary',
    hue='Gender',
    palette=['steelblue', 'tomato'],
    errorbar=None
)
plt.title('Average Salary by Department and Gender', fontsize=16)
plt.xlabel('Department')
plt.ylabel('Average Salary (Rs)')
plt.legend(title='Gender')
plt.tight_layout()
plt.show()
```

```python
# Count Plot - Frequency of each category
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

sns.countplot(data=df, x='Department', palette='Set2', ax=axes[0])
axes[0].set_title('Employee Count by Department')
axes[0].set_xlabel('Department')
axes[0].tick_params(axis='x', rotation=15)

sns.countplot(data=df, x='Performance', palette='Set3', order=['Excellent','Good','Average','Poor'], ax=axes[1])
axes[1].set_title('Performance Rating Distribution')
axes[1].set_xlabel('Performance')

plt.tight_layout()
plt.show()
```

```python
# Strip Plot - All data points
plt.figure(figsize=(11, 5))
sns.stripplot(
    data=df,
    x='Department',
    y='Salary',
    hue='Gender',
    palette=['steelblue', 'tomato'],
    jitter=True,
    alpha=0.6,
    size=4
)
plt.title('Individual Salary Points by Department', fontsize=16)
plt.legend(title='Gender')
plt.tight_layout()
plt.show()
```

---
## 4. Relationship Plots

```python
# Scatter Plot
plt.figure(figsize=(10, 6))
sns.scatterplot(
    data=df,
    x='Experience',
    y='Salary',
    hue='Department',
    size='Age',
    sizes=(30, 200),
    alpha=0.7,
    palette='Set1'
)
plt.title('Experience vs Salary (by Department & Age)', fontsize=16)
plt.xlabel('Years of Experience')
plt.ylabel('Salary (Rs)')
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

```python
# Regression Plot (scatter + trend line)
plt.figure(figsize=(9, 5))
sns.regplot(
    data=df,
    x='Experience',
    y='Salary',
    scatter_kws={'alpha': 0.4, 'color': 'steelblue'},
    line_kws={'color': 'red', 'linewidth': 2}
)
plt.title('Experience vs Salary with Trend Line', fontsize=16)
plt.xlabel('Years of Experience')
plt.ylabel('Salary (Rs)')
plt.tight_layout()
plt.show()
```

---
## 5. Heatmap — Correlation Matrix

Most important plot in data analytics — shows how variables are related.

```python
# Correlation Matrix Heatmap
numeric_cols = ['Age', 'Experience', 'Salary']
corr_matrix = df[numeric_cols].corr()

plt.figure(figsize=(8, 6))
sns.heatmap(
    corr_matrix,
    annot=True,
    fmt='.2f',
    cmap='coolwarm',
    vmin=-1, vmax=1,
    linewidths=0.5,
    square=True
)
plt.title('Correlation Matrix', fontsize=16)
plt.tight_layout()
plt.show()
print('Values close to 1 = Strong positive correlation')
print('Values close to -1 = Strong negative correlation')
print('Values close to 0 = No correlation')
```

```python
# Monthly Sales Heatmap (pivot)
np.random.seed(10)
heatmap_data = pd.DataFrame(
    np.random.randint(50, 200, (4, 12)),
    index=['North', 'South', 'East', 'West'],
    columns=['Jan','Feb','Mar','Apr','May','Jun','Jul','Aug','Sep','Oct','Nov','Dec']
)

plt.figure(figsize=(14, 5))
sns.heatmap(
    heatmap_data,
    annot=True,
    fmt='d',
    cmap='YlOrRd',
    linewidths=0.3
)
plt.title('Monthly Sales by Region (Units)', fontsize=16)
plt.tight_layout()
plt.show()
```

---
## 6. Pair Plot — All relationships at once

```python
# Pair plot for numeric columns (sample for speed)
sample_df = df[['Age', 'Experience', 'Salary', 'Department']].sample(80, random_state=42)

pair_plot = sns.pairplot(
    sample_df,
    hue='Department',
    palette='Set1',
    diag_kind='kde',
    plot_kws={'alpha': 0.5}
)
pair_plot.fig.suptitle('Pair Plot: Age, Experience, Salary by Dept', y=1.02, fontsize=15)
plt.show()
```

---
## 7. FacetGrid — Same chart for each category

```python
g = sns.FacetGrid(df, col='Department', col_wrap=3, height=4)
g.map(sns.histplot, 'Salary', bins=15, color='steelblue', edgecolor='white')
g.set_titles(col_template='{col_name}')
g.fig.suptitle('Salary Distribution by Department', y=1.02, fontsize=16)
plt.tight_layout()
plt.show()
```

## Summary — Seaborn vs Matplotlib

| Task | Matplotlib | Seaborn |
|------|-----------|----------|
| Distribution | `plt.hist()` | `sns.histplot()` |
| Box plot | `plt.boxplot()` | `sns.boxplot()` |
| Scatter | `plt.scatter()` | `sns.scatterplot()` |
| Bar | `plt.bar()` | `sns.barplot()` |
| Heatmap | Manual | `sns.heatmap()` |
| Pair plot | Manual | `sns.pairplot()` |
| Regression | Manual | `sns.regplot()` |

**Use Seaborn for:** Statistical plots, grouped analysis, quick beautiful charts

**Use Matplotlib for:** Full customization, subplots control, saving figures
