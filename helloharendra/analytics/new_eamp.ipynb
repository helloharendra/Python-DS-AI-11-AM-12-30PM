```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.DataFrame({
    'ID': [1, 2, 3],
    'Name': ['Alice', 'Bob', 'Charlie'],
    'Score': [85, 92, 78]
})

# Line Chart
df.plot(
    kind='bar',
    x='Name',
    y='Score',
    title='Scores by Student'
)
```

```python
df.plot(
    kind='pie',
    y='Score',
    labels=df['Name'],
    autopct='%1.1f%%',
    explode=[0.1, 0, 0],
    ylabel='',
    title='Scores Distribution'
)

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.DataFrame({
    'ID': [1, 2, 3],
    'Name': ['Alice', 'Bob', 'Charlie'],
    'Score': [85, 92, 78]
})


# Pie Chart
df.plot(
    kind='pie',
    y='Score',
    labels=df['Name'],      # Show student names
    autopct='%1.1f%%',
    explode=[0.1, 0, 0],
    ylabel='',
    title='Scores Distribution'
)

plt.show()
```
