```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

plt.rcParams['figure.figsize'] = (8, 5)  # Default size
plt.rcParams['font.size'] = 12
print('Libraries loaded!')
```

---
## 1. Line Plot

Best for: Trends over time (sales, temperature, stock prices)

```python
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun',
          'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']
sales  = [45, 52, 48, 61, 70, 65, 80, 75, 68, 72, 85, 90]

plt.figure(figsize=(10, 5))
plt.plot(months, sales,
         color='steelblue',
         linewidth=2,
         marker='o',
         markersize=7,
         linestyle='-',
         label='Sales 2024')

plt.title('Monthly Sales 2024', fontsize=16, fontweight='bold')
plt.xlabel('Month', fontsize=13)
plt.ylabel('Sales (in lakhs)', fontsize=13)
plt.legend(fontsize=12)
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

```python
# Multiple lines on same chart
sales_2023 = [38, 42, 40, 55, 60, 58, 72, 65, 60, 65, 75, 80]
sales_2024 = [45, 52, 48, 61, 70, 65, 80, 75, 68, 72, 85, 90]

plt.figure(figsize=(10, 5))
plt.plot(months, sales_2023, 'b-o', label='2023', linewidth=2)
plt.plot(months, sales_2024, 'r-s', label='2024', linewidth=2)

plt.title('Sales Comparison: 2023 vs 2024', fontsize=16)
plt.xlabel('Month')
plt.ylabel('Sales (Lakhs)')
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

---
## 2. Bar Chart

Best for: Comparing categories (products, departments, cities)

```python
# Vertical Bar Chart
products = ['Laptop', 'Phone', 'Tablet', 'Watch', 'Earbuds']
revenue  = [85000, 72000, 45000, 32000, 28000]

plt.figure(figsize=(9, 5))
bars = plt.bar(products, revenue,
               color=['steelblue','tomato','goldenrod','mediumseagreen','orchid'],
               edgecolor='black',
               linewidth=0.8)

# Add values on top of bars
for bar, val in zip(bars, revenue):
    plt.text(bar.get_x() + bar.get_width()/2,
             bar.get_height() + 500,
             f'Rs {val:,}',
             ha='center', va='bottom', fontsize=10)

plt.title('Product Revenue Comparison', fontsize=16)
plt.xlabel('Product')
plt.ylabel('Revenue (Rs)')
plt.grid(axis='y', alpha=0.4)
plt.tight_layout()
plt.show()
```

```python
# Horizontal Bar Chart
depts    = ['IT', 'HR', 'Finance', 'Marketing', 'Operations']
budget   = [500000, 200000, 350000, 280000, 420000]

plt.figure(figsize=(9, 5))
plt.barh(depts, budget, color='steelblue', edgecolor='black')

for i, val in enumerate(budget):
    plt.text(val + 5000, i, f'Rs {val:,}', va='center', fontsize=10)

plt.title('Department Budget Allocation', fontsize=16)
plt.xlabel('Budget (Rs)')
plt.ylabel('Department')
plt.grid(axis='x', alpha=0.4)
plt.tight_layout()
plt.show()
```

```python
# Grouped Bar Chart
quarters = ['Q1', 'Q2', 'Q3', 'Q4']
product_a = [45, 60, 55, 80]
product_b = [35, 50, 65, 70]

x = np.arange(len(quarters))
width = 0.35

fig, ax = plt.subplots(figsize=(9, 5))
bars1 = ax.bar(x - width/2, product_a, width, label='Product A', color='steelblue')
bars2 = ax.bar(x + width/2, product_b, width, label='Product B', color='tomato')

ax.set_title('Quarterly Sales by Product', fontsize=16)
ax.set_xlabel('Quarter')
ax.set_ylabel('Sales (Units)')
ax.set_xticks(x)
ax.set_xticklabels(quarters)
ax.legend()
ax.grid(axis='y', alpha=0.4)
plt.tight_layout()
plt.show()
```

```python
# Stacked Bar Chart
categories = ['North', 'South', 'East', 'West']
online     = [40, 30, 55, 35]
offline    = [60, 70, 45, 65]

plt.figure(figsize=(9, 5))
plt.bar(categories, online,  label='Online',  color='steelblue')
plt.bar(categories, offline, label='Offline', color='tomato', bottom=online)

plt.title('Sales Channel by Region', fontsize=16)
plt.xlabel('Region')
plt.ylabel('Sales (%)')
plt.legend()
plt.grid(axis='y', alpha=0.4)
plt.tight_layout()
plt.show()
```

## 3. Pie Chart

Best for: Parts of a whole (market share, budget allocation)

```python
labels  = ['Salary', 'Rent', 'Food', 'Transport', 'Entertainment', 'Savings']
values  = [50, 20, 15, 5, 5, 5]
explode = (0, 0.1, 0, 0, 0, 0.1)
colors  = ['#4472C4', '#ED7D31', '#A9D18E', '#FF0000', '#FFC000', '#70AD47']

plt.figure(figsize=(8, 8))
plt.pie(
    values,
    labels=labels,
    autopct='%1.1f%%',
    explode=explode,
    colors=colors,
    startangle=140,
    shadow=True
)
plt.title('Monthly Expense Breakdown', fontsize=16)
plt.tight_layout()
plt.show()
```

---
## 4. Scatter Plot

Best for: Relationship between two variables (height vs weight, salary vs experience)

```python
np.random.seed(42)
experience = np.random.randint(1, 15, 50)
salary     = experience * 8000 + np.random.normal(0, 15000, 50)

plt.figure(figsize=(9, 5))
plt.scatter(experience, salary,
            color='steelblue',
            alpha=0.7,
            s=80,
            edgecolors='black',
            linewidth=0.5)

# Add trend line
z = np.polyfit(experience, salary, 1)
p = np.poly1d(z)
plt.plot(np.sort(experience), p(np.sort(experience)), 'r--', linewidth=2, label='Trend')

plt.title('Experience vs Salary', fontsize=16)
plt.xlabel('Years of Experience')
plt.ylabel('Salary (Rs)')
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

---
## 5. Histogram

Best for: Distribution of data (marks, ages, salaries)

```python
np.random.seed(10)
class_marks = np.random.normal(72, 12, 200)

plt.figure(figsize=(9, 5))
plt.hist(class_marks, bins=20, color='steelblue', edgecolor='black', alpha=0.8)

plt.axvline(class_marks.mean(), color='red', linestyle='--', linewidth=2, label=f'Mean: {class_marks.mean():.1f}')
plt.axvline(np.median(class_marks), color='orange', linestyle='--', linewidth=2, label=f'Median: {np.median(class_marks):.1f}')

plt.title('Student Marks Distribution', fontsize=16)
plt.xlabel('Marks')
plt.ylabel('Number of Students')
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

---
## 6. Box Plot

Best for: Showing median, quartiles, and outliers

```python
np.random.seed(5)
it_sal      = np.random.normal(80000, 15000, 50)
finance_sal = np.random.normal(65000, 10000, 50)
hr_sal      = np.random.normal(50000, 8000, 50)

data   = [it_sal, finance_sal, hr_sal]
labels = ['IT', 'Finance', 'HR']

plt.figure(figsize=(9, 5))
bp = plt.boxplot(data, labels=labels, patch_artist=True, notch=False)

colors = ['steelblue', 'tomato', 'goldenrod']
for patch, color in zip(bp['boxes'], colors):
    patch.set_facecolor(color)
    patch.set_alpha(0.7)

plt.title('Salary Distribution by Department', fontsize=16)
plt.xlabel('Department')
plt.ylabel('Annual Salary (Rs)')
plt.grid(axis='y', alpha=0.4)
plt.tight_layout()
plt.show()
```

## 7. Subplots — Multiple Charts in One Figure

```python
fig, axes = plt.subplots(2, 2, figsize=(14, 10))
fig.suptitle('Sales Dashboard', fontsize=20, fontweight='bold')

# Chart 1: Line
months_short = ['J','F','M','A','M','J','J','A','S','O','N','D']
sales_data = [45,52,48,61,70,65,80,75,68,72,85,90]
axes[0, 0].plot(months_short, sales_data, 'b-o', linewidth=2)
axes[0, 0].set_title('Monthly Sales Trend')
axes[0, 0].set_xlabel('Month')
axes[0, 0].set_ylabel('Sales')
axes[0, 0].grid(True, alpha=0.3)

# Chart 2: Bar
regions  = ['North', 'South', 'East', 'West']
rev      = [85, 60, 72, 55]
axes[0, 1].bar(regions, rev, color=['#4472C4','#ED7D31','#A9D18E','#FFC000'])
axes[0, 1].set_title('Revenue by Region')
axes[0, 1].set_ylabel('Revenue (Lakhs)')

# Chart 3: Pie
cats   = ['Electronics', 'Clothing', 'Food', 'Others']
shares = [40, 25, 20, 15]
axes[1, 0].pie(shares, labels=cats, autopct='%1.1f%%', startangle=90)
axes[1, 0].set_title('Category Share')

# Chart 4: Histogram
np.random.seed(42)
orders = np.random.normal(5000, 1500, 200)
axes[1, 1].hist(orders, bins=20, color='steelblue', edgecolor='black', alpha=0.8)
axes[1, 1].set_title('Order Value Distribution')
axes[1, 1].set_xlabel('Order Value (Rs)')
axes[1, 1].set_ylabel('Count')

plt.tight_layout()
plt.show()
```

## 8. Area Plot & Fill Between

```python
weeks  = list(range(1, 25))
revenue  = [40, 45, 42, 55, 60, 58, 65, 70, 68, 72, 78, 80,
            75, 82, 85, 90, 88, 92, 95, 100, 98, 105, 102, 110]
costs  = [30, 32, 30, 38, 40, 38, 42, 45, 44, 47, 50, 51,
          49, 52, 54, 56, 55, 58, 60, 62, 61, 64, 63, 66]

plt.figure(figsize=(11, 5))
plt.plot(weeks, revenue, 'g-', linewidth=2, label='Revenue')
plt.plot(weeks, costs, 'r-', linewidth=2, label='Cost')
plt.fill_between(weeks, revenue, costs, alpha=0.2, color='green', label='Profit Zone')
plt.fill_between(weeks, costs, 0, alpha=0.1, color='red', label='Cost Zone')

plt.title('Revenue vs Cost (Profit Analysis)', fontsize=16)
plt.xlabel('Week')
plt.ylabel('Amount (Lakhs)')
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

## 10. Real-World Project: Complete Sales Report

```python
# Create sample sales data
np.random.seed(42)
df = pd.DataFrame({
    'Month': ['Jan','Feb','Mar','Apr','May','Jun','Jul','Aug','Sep','Oct','Nov','Dec'],
    'Online':  np.random.randint(50, 120, 12),
    'Offline': np.random.randint(30, 100, 12),
    'Returns': np.random.randint(5, 25, 12)
})

df['Total'] = df['Online'] + df['Offline']
df['Profit'] = df['Total'] - df['Returns']

fig, axes = plt.subplots(2, 2, figsize=(14, 10))
fig.suptitle('Annual Sales Report 2024', fontsize=22, fontweight='bold', y=1.01)

# 1. Monthly total trend
axes[0,0].plot(df['Month'], df['Total'], 'b-o', linewidth=2, label='Total Sales')
axes[0,0].plot(df['Month'], df['Profit'], 'g-s', linewidth=2, label='Net Profit')
axes[0,0].set_title('Sales & Profit Trend')
axes[0,0].legend()
axes[0,0].tick_params(axis='x', rotation=45)
axes[0,0].grid(True, alpha=0.3)

# 2. Online vs Offline stacked bar
x = np.arange(len(df['Month']))
axes[0,1].bar(x, df['Online'], label='Online', color='#4472C4')
axes[0,1].bar(x, df['Offline'], bottom=df['Online'], label='Offline', color='#ED7D31')
axes[0,1].set_title('Online vs Offline Sales')
axes[0,1].set_xticks(x)
axes[0,1].set_xticklabels(df['Month'], rotation=45)
axes[0,1].legend()

# 3. Returns histogram
axes[1,0].bar(df['Month'], df['Returns'], color='tomato', edgecolor='black')
axes[1,0].set_title('Monthly Returns')
axes[1,0].tick_params(axis='x', rotation=45)
axes[1,0].set_ylabel('Returns (Units)')

# 4. Channel distribution pie
total_online  = df['Online'].sum()
total_offline = df['Offline'].sum()
axes[1,1].pie([total_online, total_offline],
              labels=['Online', 'Offline'],
              autopct='%1.1f%%',
              colors=['#4472C4', '#ED7D31'],
              startangle=90, shadow=True)
axes[1,1].set_title(f'Sales Channel Split\n(Total: {total_online+total_offline} units)')

plt.tight_layout()
plt.show()

print(f'Best month: {df.loc[df["Profit"].idxmax(), "Month"]} (Profit: {df["Profit"].max()})')
print(f'Total Annual Sales: {df["Total"].sum()} units')
```

---
## Summary — Chart Selection Guide

| Chart Type | When to Use |
|------------|-------------|
| **Line** | Trends over time |
| **Bar** | Compare categories |
| **Pie** | Parts of a whole |
| **Scatter** | Relationship between two variables |
| **Histogram** | Distribution of data |
| **Box Plot** | Spread, median, outliers |
| **Subplots** | Multiple charts in one figure |
| **Area** | Cumulative trends |
