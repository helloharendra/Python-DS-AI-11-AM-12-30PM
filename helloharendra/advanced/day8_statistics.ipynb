## Measures of Central Tendency
Before doing any deeper analysis, you usually want a single number that summarizes "where the data sits" — mean, median, and mode are the three standard ways to describe that center, and each behaves differently in the presence of skew or repeated values. The mean is the arithmetic average (sensitive to extreme values), the median is the middle value when data is sorted (more robust to outliers), and the mode is the most frequently occurring value (useful for categorical or repeated data). In the output, the mean (36.6) and median (39.5) differ slightly because of a couple of low outliers (12, 22) pulling the average down, and the mode result shows `45` occurs most often (3 times) in this dataset.

```python
import numpy as np
from scipy import stats

data = [23, 45, 12, 67, 34, 45, 22, 45]
print('Mean:', np.mean(data))    # arithmetic average of all values
print('Median:', np.median(data))  # middle value when sorted (average of two middles here, since n is even)
print('Mode:', stats.mode(data, keepdims=True))  # most frequently occurring value and its count
```

## Measures of Dispersion
Knowing the "center" of your data isn't enough — you also need to know how spread out it is, since two datasets can have the same mean but very different amounts of variability, which matters a lot for risk assessment, quality control, and model reliability. Variance measures the average squared distance of each point from the mean, standard deviation is variance's square root (back in the original units, so it's easier to interpret), and range is simply the difference between the maximum and minimum values. In the output, the standard deviation (~16.4) tells you values typically deviate from the mean of 36.6 by about that much, and the range (55) shows the full spread from the smallest to largest value.

```python
print('Variance:', np.var(data))   # average of squared deviations from the mean
print('Std Dev:', np.std(data))    # square root of variance, in the same units as the data
print('Range:', max(data) - min(data))  # difference between the largest and smallest values
```

## Correlation
Correlation is used constantly to check whether two numeric variables are related before assuming one causes or predicts the other — for example, whether hours studied relates to exam score. `np.corrcoef(x, y)` computes the full correlation matrix between `x` and `y`, and since the matrix is symmetric with 1.0 on the diagonal (each variable perfectly correlates with itself), `[0, 1]` picks out the actual x-vs-y correlation value. The output value of ~0.77 is a strong positive correlation, meaning as `x` increases, `y` tends to increase too — but remember correlation only measures a linear relationship, not causation.

```python
x = [1, 2, 3, 4, 5]
y = [2, 4, 5, 4, 5]
# corrcoef returns a 2x2 matrix; [0,1] pulls out the x-vs-y correlation coefficient
print('Correlation:', np.corrcoef(x, y)[0, 1])
```

## Normal Distribution
The normal ("bell curve") distribution shows up throughout real-world data — heights, measurement errors, test scores — and many statistical tests and machine learning methods assume data is roughly normally distributed, so recognizing this shape is a core skill. `np.random.normal(loc=50, scale=10, size=1000)` generates 1000 random samples from a normal distribution centered at `loc=50` (the mean) with `scale=10` (the standard deviation, controlling spread), simulating what real normally-distributed data looks like. The histogram in the output should show the familiar symmetric bell shape, peaking around 50 and tapering off toward both sides as values get farther from the mean.

```python
import matplotlib.pyplot as plt
# loc = mean (center), scale = standard deviation (spread), size = number of samples to draw
samples = np.random.normal(loc=50, scale=10, size=1000)
plt.hist(samples, bins=30)
plt.title('Normal Distribution')
plt.show()
```

## Hypothesis Testing (t-test)
Hypothesis testing is how you decide, with statistical rigor, whether an observed difference between two groups is real or could just be due to random chance — for example, whether group B genuinely scores higher than group A, or whether that's a coincidence of small sample sizes. An independent two-sample t-test compares the means of two groups and returns a t-statistic (how many standard errors apart the means are) and a p-value (the probability of seeing a difference this large if there were truly no difference between groups). In the output, the p-value (~0.003) is well below the common 0.05 significance threshold, so we'd conclude the difference between group_a and group_b is statistically significant — the negative t-statistic simply reflects that group_a's mean is lower than group_b's.

```python
group_a = [23, 25, 28, 22, 27]
group_b = [30, 32, 29, 35, 31]
# Independent t-test: compares the means of two separate groups
t_stat, p_value = stats.ttest_ind(group_a, group_b)
print('t-statistic:', t_stat)   # how many standard errors apart the two group means are
print('p-value:', p_value)      # probability of this difference occurring by chance alone
```
