## The .str Accessor
Text columns are extremely common in real datasets — names, emails, addresses, free-form notes — and they usually need cleaning before they're usable. Looping over rows in Python to clean strings one at a time is slow and unpythonic; pandas' `.str` accessor gives you vectorized versions of Python's normal string methods (`.lower()`, `.strip()`, `.contains()`, etc.) that apply to an entire column at once, the same way arithmetic operators apply to a whole numeric column. This is the standard entry point for any text-cleaning work in pandas. Below, `.str.lower()` lowercases every email in the column in one call — the output shows all addresses now in consistent lowercase, which matters because "Alice@Gmail.com" and "alice@gmail.com" should usually be treated as the same value.</cell id="cell-1">


```python
import pandas as pd

df = pd.DataFrame({'Email': ['Alice@Gmail.com', 'bob@yahoo.com', 'CHARLIE@OUTLOOK.COM']})
df['Email'] = df['Email'].str.lower()  # vectorized lowercase applied to every row at once
print(df)
```

## Searching Text
A common task is filtering rows to only those whose text matches some pattern — finding all customers with a Gmail address, all log lines containing "ERROR", and so on. `.str.contains()` returns a boolean Series (True/False for each row, based on whether the pattern is found), which you can then use just like any other boolean mask to filter the DataFrame with `df[...]`. This is the text equivalent of a numeric filter like `df[df['age'] > 30]`. The output here is a DataFrame containing only the rows where the Email column contains "gmail" — in this case, just Alice's row.</cell id="cell-3">


```python
# .str.contains returns a boolean mask (True/False per row); use it to filter the DataFrame
print(df[df['Email'].str.contains('gmail')])
```

## Extracting with Regex
Sometimes you don't just want to check if a pattern exists — you want to pull a specific piece out of the text, like the domain from an email or the numeric part of a product code. `.str.extract()` takes a regular expression with a capture group (the part in parentheses) and returns just what that group matched, as a new column. Regex is the standard tool for this kind of pattern-based text extraction across data science and software engineering generally, so it's worth getting comfortable with. Here, the pattern `@(.+)` matches an "@" followed by everything after it, and the captured group (everything after the @) becomes the new `Domain` column — so "alice@gmail.com" produces "gmail.com".</cell id="cell-5">


```python
# @(.+) captures everything after the '@' symbol; str.extract pulls that capture group into a new column
df['Domain'] = df['Email'].str.extract(r'@(.+)')
print(df)
```

## Splitting & Replacing
Two more everyday text operations: breaking one column into several parts, and swapping out unwanted patterns. `.str.split(sep, expand=True)` splits each string on a separator and spreads the pieces into separate new columns (instead of returning a list per row), which is handy for something like splitting "first last" names or, here, splitting an email into its username and host. `.str.replace()` with `regex=True` finds a pattern anywhere in the string and swaps it for something else — useful for standardizing formats or redacting/correcting text at scale. In the output, `User`/`Host` are the two halves of each email split on "@", and the replace call returns a new Series with ".com" swapped for ".in" (note: this only prints the result, it doesn't modify `df['Email']` itself).</cell id="cell-7">


```python
# split each email on '@' and expand the two pieces into separate User/Host columns
df[['User', 'Host']] = df['Email'].str.split('@', expand=True)
print(df)
# regex replace: swap '.com' for '.in' in each email (returns a new Series, doesn't modify df['Email'])
print(df['Email'].str.replace(r'\.com', '.in', regex=True))
```
