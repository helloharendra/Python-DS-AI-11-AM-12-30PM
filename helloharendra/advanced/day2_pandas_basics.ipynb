## Pandas Series
A Series is pandas' 1D building block — think of it as a single labeled column of data, similar to a NumPy array but with an explicit index attached to each value instead of just a position number. This labeling matters in real work because it lets you look up values by a meaningful name (like `'a'` or a date or a customer ID) instead of remembering numeric positions, and it's what every column of a DataFrame actually is under the hood. The code creates a Series from a list of numbers and gives it custom labels `'a'`, `'b'`, `'c'`, `'d'` instead of the default 0, 1, 2, 3 — when printed, you'll see each label next to its value, plus the data type (`dtype`) of the values at the bottom.

```python
import pandas as pd

# Create a Series with custom labels instead of default 0,1,2,3 positions
s = pd.Series([10, 20, 30, 40], index=['a', 'b', 'c', 'd'])
print(s)
```

## Creating a DataFrame
A DataFrame is the main structure you'll use in almost every pandas task — a 2D labeled table with rows and columns, exactly like a spreadsheet or a database table, where each column is actually a Series. Most real analysis starts by getting your data into a DataFrame, either by loading a file or, as shown here, building one directly from a Python dictionary where each key becomes a column name and each value list becomes that column's data. When you print the result, pandas automatically adds a numeric row index (0, 1, 2) on the left and aligns each column underneath its header, giving you a clean tabular view of the data.

```python
!pip install pandas
```

```python
# Build a DataFrame from a dictionary: keys become column names, lists become column values
data = {
    'Name': ['Alice', 'Bob', 'Charlie'],
    'Age': [25, 30, 35],
    'City': ['NYC', 'LA', 'Chicago']
}
df = pd.DataFrame(data)
print(df)
```

## Reading a CSV File
In practice you almost never type data in by hand — it comes from CSV files, Excel sheets, or databases, so knowing how to write and read CSVs is a core everyday skill. `to_csv()` saves a DataFrame to disk as a plain-text comma-separated file, and `read_csv()` loads a CSV back into a DataFrame; `index=False` is used here so pandas doesn't write the row numbers as an extra column in the file. Running this cell writes `students.csv` to disk, reloads it into `df_loaded`, and prints it — the output should look identical to the original DataFrame, confirming the round-trip worked correctly.

```python
# index=False avoids writing pandas' row numbers as an extra unnamed column in the CSV
df.to_csv('students.csv', index=False)
df_loaded = pd.read_csv('students.csv')  # read the CSV back into a fresh DataFrame
print(df_loaded)
```

## Exploring a DataFrame
Before doing any analysis, you need a quick sense of what a dataset looks like — its size, column types, missing values, and the general range of the numbers — and this handful of methods is exactly the "first look" toolkit every data scientist reaches for. `head()`/`tail()` preview the first/last few rows so you're not printing an entire (potentially huge) dataset, `info()` shows column names, data types, and non-null counts (useful for spotting missing data), `describe()` gives summary statistics for numeric columns, and `shape` tells you the (rows, columns) size. Reading the output: `info()` confirms all 3 rows have non-null values in every column, and `describe()`'s rows (count, mean, std, min, 25%, 50%, 75%, max) summarize the distribution of the `Age` column.

```python
print(df.head(2))     # first 2 rows — a quick peek at the top of the data
print(df.tail(2))     # last 2 rows — a quick peek at the bottom of the data
print(df.info())      # column names, dtypes, and non-null counts (spot missing data at a glance)
print(df.describe())  # summary statistics (count, mean, std, min/max, quartiles) for numeric columns
print(df.shape)       # (number of rows, number of columns)
```

## Selecting Columns & Rows
Pulling out specific columns or rows is something you'll do constantly — to inspect a single feature, build a subset for modeling, or check a particular record. Selecting a single column with `df['col']` returns a Series, while a list of column names `df[['col1', 'col2']]` returns a smaller DataFrame; for rows, `.iloc[]` selects by integer position (like list indexing) and `.loc[]` selects by the row's label (which happens to also be integers here since we didn't set a custom index). The output shows each selection style side by side: a single-column Series, a two-column DataFrame, and two different ways for extracting one row as a Series.

```python
print(df['Name'])          # single column -> returns a Series
print(df[['Name', 'Age']]) # multiple columns (note the double brackets) -> returns a DataFrame
print(df.iloc[0])          # first row by integer position
print(df.loc[1])           # row labeled 1 (label-based lookup)
```

## Filtering Rows
Filtering is how you answer questions like "show me only the customers over 30" or "which transactions failed" — it's one of the most-used operations in any data analysis. `df['Age'] > 27` produces a Series of `True`/`False` values (one per row), and passing that boolean Series back into `df[...]` keeps only the rows where the condition is `True` — this pattern is called boolean indexing. The output shows only the rows where `Age` is greater than 27 (Bob and Charlie), while Alice (age 25) is filtered out.

```python
# df['Age'] > 27 creates a True/False mask; df[mask] keeps only the rows where it's True
print(df[df['Age'] > 27])
```
