## Text Preprocessing
Machine learning models work with numbers, not raw sentences, so text always needs cleaning and structuring before it can be used as input. Preprocessing typically means lowercasing (so "Data" and "data" aren't treated as different words), stripping out punctuation and other noise, and tokenizing — splitting the text into individual words ("tokens") that can later be counted or vectorized. This is the first step of virtually any NLP pipeline, whether you're building a spam filter, a sentiment classifier, or a search engine. Below, a regex strips out anything that isn't a letter or whitespace, `.lower()` normalizes case, and `.split()` breaks the cleaned string into a list of individual word tokens — the printed output is that list.</cell id="cell-1">


```python
import re

text = "Data Science is AMAZING! Isn't it?"
clean = re.sub(r'[^a-zA-Z\s]', '', text).lower()  # keep only letters/spaces, then lowercase everything
tokens = clean.split()  # split the cleaned string into a list of word tokens
print(tokens)
```

## Removing Stopwords
Stopwords are very common words ("is", "the", "a", "it", etc.) that appear in almost every sentence and usually don't help distinguish one document's meaning from another's. Removing them reduces noise and dimensionality before modeling, which can improve both speed and accuracy for tasks like topic modeling or classification — though for some tasks (like sentiment analysis, where "not" matters) you have to be careful about what you strip out. Here we use a small hardcoded set of stopwords and keep only the tokens that aren't in it, via a list comprehension. The printed output is the same token list as before but with the low-information words filtered out.</cell id="cell-3">


```python
stopwords = {'is', 'it', 'the', 'a', 'an'}
# keep only tokens that are NOT in the stopword set
filtered = [word for word in tokens if word not in stopwords]
print(filtered)
```

## Bag of Words
Once text is cleaned, it still needs to become numbers before any ML model can use it — Bag of Words (BoW) is the simplest way to do that. It builds a vocabulary of every unique word across all documents, then represents each document as a row of counts: how many times each vocabulary word appears in it. It's called "bag of words" because word order is thrown away entirely — only the counts matter. This is a foundational technique behind text classification and search, even though more advanced methods (like TF-IDF or embeddings) usually improve on it. In the output, `get_feature_names_out()` lists the vocabulary (columns), and `bow.toarray()` shows one row per document with the count of each vocabulary word in that document.</cell id="cell-5">


```python
from sklearn.feature_extraction.text import CountVectorizer

docs = ['I love data science', 'Data science is fun', 'I love learning AI']
vectorizer = CountVectorizer()
bow = vectorizer.fit_transform(docs)  # learns the vocabulary from docs AND converts docs to count vectors
print(vectorizer.get_feature_names_out())  # the vocabulary (one entry per unique word, alphabetical)
print(bow.toarray())  # one row per document, one column per vocab word, values = word counts
```

## TF-IDF
Plain word counts have a flaw: common words that appear in every document (even after removing basic stopwords) end up with high counts without actually being informative. TF-IDF (Term Frequency - Inverse Document Frequency) fixes this by weighing each word by how frequent it is in one document (TF) multiplied by how rare it is across all documents (IDF) — so a word that appears often in one document but rarely elsewhere gets a high score, while a word common everywhere gets pulled down. This makes TF-IDF a much stronger feature representation for tasks like search ranking and document classification than raw Bag of Words. The output below shows the same vocabulary as before, but now the values are weighted scores instead of raw counts — words distinctive to a particular document will have relatively higher values in that document's row.</cell id="cell-7">


```python
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf = TfidfVectorizer()
matrix = tfidf.fit_transform(docs)  # learns vocab and computes TF-IDF weighted scores per document
print(tfidf.get_feature_names_out())
print(matrix.toarray().round(2))  # rounded for readability - higher score = more distinctive to that doc
```
