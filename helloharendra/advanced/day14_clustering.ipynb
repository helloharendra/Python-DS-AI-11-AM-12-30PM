# Day 14: Unsupervised Learning - K-Means Clustering

## Unsupervised Learning
No labels are given - the model finds hidden groups (clusters) in the data.

```python
from sklearn.datasets import load_iris

X, y = load_iris(return_X_y=True)
X = X[:, :2]  # use first 2 features for easy plotting
```

## Elbow Method
Helps choose the right number of clusters (k) by looking for the 'bend' in the plot.

```python
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

inertias = []
for k in range(1, 7):
    km = KMeans(n_clusters=k, random_state=42, n_init=10).fit(X)
    inertias.append(km.inertia_)
plt.plot(range(1, 7), inertias, marker='o')
plt.xlabel('k')
plt.ylabel('Inertia')
plt.title('Elbow Method')
plt.show()
```

## Fit K-Means
Group the data into k clusters based on similarity.

```python
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
labels = kmeans.fit_predict(X)
print(labels[:10])
```

## Visualize Clusters
Plot the data colored by the cluster each point was assigned to.

```python
plt.scatter(X[:, 0], X[:, 1], c=labels, cmap='viridis')
plt.scatter(kmeans.cluster_centers_[:, 0], kmeans.cluster_centers_[:, 1], c='red', marker='X', s=200)
plt.title('K-Means Clusters')
plt.show()
```
