# Day 6: Statistical Visualization with Seaborn

## Setup
Seaborn is a plotting library built on top of Matplotlib that's tailored for statistical visualization — it knows how to work directly with DataFrames and understands concepts like categories and distributions, so you get informative charts (with sensible colors, spacing, and statistical overlays) with far less code than raw Matplotlib. This cell just sets up the sample dataset that every plot in this notebook will use: a small DataFrame with each employee's `Age`, `Salary`, and `Department`. There's no visible output here, but running it makes `df` available for all the plotting examples that follow.

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

# Sample employee dataset used throughout this notebook's plotting examples
df = pd.DataFrame({
    'Age': [22, 25, 47, 52, 46, 56, 23, 27],
    'Salary': [25000, 32000, 65000, 72000, 68000, 80000, 27000, 35000],
    'Department': ['IT', 'IT', 'HR', 'HR', 'Sales', 'Sales', 'IT', 'HR']
})
```

## Distribution Plot
Understanding the shape of a numeric column — is it clustered around one value, spread evenly, or split into two groups — is a key step before choosing statistics or models that assume a particular distribution (like normality). `sns.histplot(df['Salary'], kde=True)` draws a histogram (bars showing how many values fall into each range) and overlays a KDE curve (`kde=True`), a smoothed line estimating the underlying probability distribution, which makes the overall shape easier to read than bars alone. In the output, the bars show how salaries are grouped into ranges, and the smooth curve traces the general shape of that spread — here you'd expect to see two loose clusters, since salaries range from entry-level (~25k-35k) to senior (~65k-80k).

```python
# kde=True overlays a smoothed density curve on top of the histogram bars
sns.histplot(df['Salary'], kde=True)
plt.title('Salary Distribution')
plt.show()
```

## Box Plot
Box plots let you compare how a numeric value is distributed across several categories in one chart, and are especially good at surfacing outliers — a common early question is "does salary vary noticeably by department, and are there any unusual values?" Each box shows the middle 50% of values (the IQR, from Q1 to Q3), the line inside is the median, the "whiskers" extend to the typical range, and any points plotted beyond the whiskers are flagged as potential outliers. In the output, you'll see one box per department (HR, IT, Sales), letting you compare their salary ranges and medians side by side at a glance.

```python
# One box per Department, showing the median, IQR, and any outlier salaries
sns.boxplot(x='Department', y='Salary', data=df)
plt.title('Salary by Department')
plt.show()
```

## Count Plot
When you need a quick check on how many records fall into each category (is the dataset balanced across departments, classes, or groups?), a count plot is the simplest way to see it — this is especially important before training classification models, where imbalanced categories can bias results. `sns.countplot(x='Department', data=df)` automatically tallies how many rows have each unique value in the `Department` column and draws one bar per category, with height equal to the count. The output shows a bar for each department; comparing bar heights tells you at a glance whether one department has noticeably more employees than the others.

```python
# countplot tallies rows per unique Department value automatically (no manual groupby needed)
sns.countplot(x='Department', data=df)
plt.title('Employees per Department')
plt.show()
```

## Scatter & Pair Plot
Beyond looking at one variable at a time, you often need to see how multiple numeric variables relate to each other and whether those relationships differ by category — for example, does the Age-vs-Salary trend look the same across departments? `sns.scatterplot(..., hue='Department')` plots Age against Salary with points colored by department, so you can spot both the overall trend and any per-group differences in one chart; `sns.pairplot(df, hue='Department')` goes further and automatically generates a grid of scatter plots for every pair of numeric columns (plus a distribution plot on the diagonal), which is a fast way to scan an entire dataset for interesting relationships. In the output, the scatter plot shows Age vs Salary with each department in a different color, and the pair plot grid lets you compare every numeric column against every other one at once.

```python
# hue='Department' colors points by category so group patterns are visible
sns.scatterplot(x='Age', y='Salary', hue='Department', data=df)
plt.title('Age vs Salary')
plt.show()
# pairplot auto-generates scatter plots for every pair of numeric columns, colored by Department
sns.pairplot(df, hue='Department')
plt.show()
```

## Correlation Heatmap
Correlation measures how strongly two numeric variables move together (from -1, perfectly opposite, to +1, perfectly aligned), and checking correlations early helps you spot redundant features or confirm expected relationships before modeling. `df[...].corr()` computes the pairwise correlation coefficient between every pair of numeric columns and returns it as a small table, and `sns.heatmap(corr, annot=True, cmap='coolwarm')` visualizes that table as a colored grid — `annot=True` prints the actual numbers on each cell, and the `coolwarm` colormap uses warm colors for strong positive correlation and cool colors for negative. In the output, the diagonal will always show 1.0 (a variable is perfectly correlated with itself), and the Age-Salary cell shows how strongly age and salary move together in this dataset — expect a strong positive value, since older employees here tend to earn more.

```python
corr = df[['Age', 'Salary']].corr()  # pairwise correlation coefficients between columns
# annot=True prints the numeric values on each cell; coolwarm maps positive/negative to color
sns.heatmap(corr, annot=True, cmap='coolwarm')
plt.title('Correlation Heatmap')
plt.show()
```
