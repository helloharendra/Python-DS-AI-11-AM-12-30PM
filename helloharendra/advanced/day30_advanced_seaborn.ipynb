## Setup
Advanced seaborn plots are built to reveal patterns across categories and relationships between variables, so we need a dataset with more structure than a single list of numbers - here, sales figures broken down by day of the week and by store. This kind of "long format" DataFrame (one row per observation, with category columns like `day` and `store` alongside the numeric `sales` value) is exactly the shape seaborn expects, since its plotting functions work by splitting and coloring data based on column names rather than requiring you to slice the data yourself. Running this cell just builds the DataFrame in memory - there's no visible output - but every seaborn plot in the rest of this notebook draws from this same `df`.

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

# Long-format dataset: 20 rows total - 4 weeks of Mon-Fri sales,
# alternating between store A and store B in blocks of 5
df = pd.DataFrame({
    'day': ['Mon', 'Tue', 'Wed', 'Thu', 'Fri'] * 4,
    'sales': [20, 35, 30, 40, 25, 22, 38, 28, 45, 30, 18, 33, 32, 42, 27, 25, 40, 29, 44, 31],
    'store': ['A'] * 5 + ['B'] * 5 + ['A'] * 5 + ['B'] * 5
})
```

## FacetGrid
When you want to compare the same kind of chart across several categories - like "how do sales trends differ between store A and store B?" - manually creating a separate plot for each one is repetitive. `FacetGrid` automates this: you tell it which column to split on (`col='store'`), and it builds a grid of subplots, one panel per unique value in that column, sharing consistent axes for easy comparison. `g.map()` then applies the same plotting function (here `sns.lineplot`) to each panel using that panel's slice of the data. The result is two side-by-side line charts, one for store A and one for store B, both showing sales across the days of the week - letting you visually compare their patterns without overlaying and cluttering a single chart.

```python
# Build a grid with one panel per distinct value of 'store'
g = sns.FacetGrid(df, col='store')
# Draw the same lineplot (day vs sales) inside each panel, using that panel's data slice
g.map(sns.lineplot, 'day', 'sales')
plt.show()
```

## Joint Plot
A joint plot is useful when you want to see both the relationship between two numeric variables (as a scatter plot in the middle) and each variable's individual distribution (as histograms along the top and right margins) in one combined figure - saving you from making three separate plots. Normally you'd pass two different numeric columns for `x` and `y` (for example price vs. quantity) to see how they relate. This toy dataset only has one numeric column (`sales`), so the example below plots `sales` against itself purely to demonstrate the joint plot's layout - in practice you'd always compare two distinct variables. Even here, notice the structure: a scatter/point cloud in the center, with a histogram of the same variable's spread mirrored on each margin.

```python
# kind='scatter' draws points in the center plus a histogram of each axis on the margins
# (using 'sales' for both axes here only because this dataset has one numeric column)
sns.jointplot(x='sales', y='sales', data=df, kind='scatter')
plt.show()
```

## Violin & Swarm Plot
Both plots compare a numeric variable across categories, but they emphasize different things and are often used together. A violin plot mirrors a box plot but replaces the box with a smoothed, mirrored density curve, so its width at any height shows how common values around that height are - wider sections mean more data points cluster there, making it easy to spot whether a distribution is skewed, bimodal, or tightly concentrated. A swarm plot instead shows every single data point as a dot, nudged sideways so points don't overlap - useful for small datasets where you want to see the raw data rather than a summarized shape. Running both back to back on the same `store` vs `sales` comparison lets you cross-check: does the violin's "fat" region actually correspond to a dense cluster of real points in the swarm?

```python
# Violin plot: shows the smoothed distribution shape of sales for each store
sns.violinplot(x='store', y='sales', data=df)
plt.title('Sales Distribution by Store')
plt.show()
# Swarm plot: shows every individual sales value as a point, offset to avoid overlap
sns.swarmplot(x='store', y='sales', data=df)
plt.title('Individual Sales Points')
plt.show()
```
