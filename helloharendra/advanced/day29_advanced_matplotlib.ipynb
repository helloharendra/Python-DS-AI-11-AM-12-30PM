## Annotations
When presenting a chart to stakeholders, it's often not enough to just show the data - you want to draw attention to a specific point, like a peak, a dip, or an anomaly, and explain why it matters. Matplotlib's `annotate` function lets you place text and an arrow pointing at any coordinate on the plot, turning a plain line chart into a mini-story. In the code below, we plot a simple sales line and then annotate the highest point (4, 32) with the label "Peak", placing the text slightly offset so it doesn't overlap the point, with an arrow connecting the label back to the exact data point. The result is a chart where the eye is immediately drawn to the point being called out, rather than making the viewer hunt for it.

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [10, 25, 18, 32, 28]
plt.plot(x, y, marker='o')
# Annotate the peak: xy is the point being pointed at, xytext is where the
# label text sits, and arrowprops draws an arrow connecting the two
plt.annotate('Peak', xy=(4, 32), xytext=(3, 33), arrowprops={'arrowstyle': '->'})
plt.title('Sales with Annotation')
plt.show()
```

## Styling with Themes
Matplotlib's default look is functional but plain, and manually setting colors, fonts, and backgrounds for every chart is tedious. Built-in style sheets bundle up a whole set of visual defaults - background color, gridlines, color cycle, fonts - so calling `plt.style.use()` once instantly changes the appearance of every chart drawn afterward, without touching any plotting code. This is useful for quickly giving a presentation or report a consistent, polished look. Below we switch to the `'seaborn-v0_8-darkgrid'` style (a dark-gridded background borrowed from seaborn), draw a plot, and then switch back to `'default'` so it doesn't affect any charts made later in the notebook - it's good practice to reset the style after a one-off demo like this.

```python
# Apply a built-in style sheet - changes background, gridlines, and colors globally
plt.style.use('seaborn-v0_8-darkgrid')
plt.plot(x, y)
plt.title('Styled Plot')
plt.show()
# Reset to matplotlib's default style so later plots aren't affected
plt.style.use('default')
```

## Dual Y-Axes
Sometimes you want to compare two metrics that move together but live on completely different scales - for example sales in the thousands versus a returns count in single digits - and plotting them on one shared y-axis would squash one of them flat. A dual-axis chart solves this by giving each line its own y-axis: one on the left, one on the right, while sharing the same x-axis. Matplotlib does this through `ax.twinx()`, which creates a second axes object overlaid on the first but with an independent y-scale. Below, `ax1` plots sales in blue against the left axis, and `ax2` (created via `twinx()`) plots returns in red against the right axis; coloring each axis label to match its line makes it clear at a glance which scale belongs to which series.

```python
fig, ax1 = plt.subplots()
# Left y-axis: sales, plotted in blue
ax1.plot(x, y, color='blue')
ax1.set_ylabel('Sales', color='blue')
# twinx() creates a second axes sharing the same x-axis but with its own y-scale
ax2 = ax1.twinx()
# Right y-axis: returns, plotted in red on its own independent scale
ax2.plot(x, [5, 4, 6, 3, 7], color='red')
ax2.set_ylabel('Returns', color='red')
plt.title('Sales vs Returns')
plt.show()
```

## Custom Ticks & Grid
By default matplotlib picks axis tick positions and labels automatically based on the numeric data, but often you want the axis to show meaningful labels instead - like day names instead of raw numbers 1 through 5. `plt.xticks()` lets you override both the tick positions and the text shown at each position, which is essential when your x-axis represents categories (like weekdays) rather than a continuous number line. Adding a dashed, semi-transparent grid with `plt.grid()` also makes it easier for viewers to read off approximate values without cluttering the chart. In the code below, the numeric x-values `[1,2,3,4,5]` are relabeled as weekday names, and a light dashed grid is layered behind the line for easier reading.

```python
plt.plot(x, y)
# Replace the numeric x-axis labels (1-5) with weekday names at those same positions
plt.xticks(x, ['Mon', 'Tue', 'Wed', 'Thu', 'Fri'])
# Add a light dashed grid to make values easier to read off the chart
plt.grid(True, linestyle='--', alpha=0.5)
plt.title('Custom Ticks')
plt.show()
```
