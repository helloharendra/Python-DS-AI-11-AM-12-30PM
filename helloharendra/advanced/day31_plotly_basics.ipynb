```python
# pip install plotly
import plotly.express as px
import pandas as pd
```

## Interactive Line Chart
`plotly.express` (imported as `px`) is a high-level API, much like seaborn, where you pass a DataFrame plus column names and it builds the chart for you. The key difference from matplotlib/seaborn is that the output isn't a static image - it's an interactive figure where you can hover over any point to see its exact value, click-and-drag to zoom into a region, and pan around, all without writing any extra code. This makes it well suited for exploratory analysis, where you want to inspect specific values rather than just eyeball a trend. Below, a simple line chart of daily sales is created with `px.line`, mapping `day` to the x-axis and `sales` to the y-axis; calling `fig.show()` renders the interactive chart.

```python
# Simple day-by-day sales series
df = pd.DataFrame({'day': range(1, 8), 'sales': [20, 35, 30, 40, 25, 45, 38]})
# px.line builds an interactive line chart directly from column names
fig = px.line(df, x='day', y='sales', title='Daily Sales (Interactive)')
fig.show()
```

## Interactive Scatter Plot
A scatter plot is the standard way to explore the relationship between two numeric variables, and plotly's version adds automatic hover tooltips showing the exact values behind each point - useful when you want to identify a specific outlier or data point without cross-referencing a table. `px.scatter` also supports mapping extra columns to visual channels beyond just x and y: `color` assigns a different color per category (here, department), and `size` scales each point's size by a numeric value (here, salary), letting you encode up to four dimensions (x, y, color, size) in a single 2D chart. In the example below, each point is an employee, positioned by age and salary, colored by department, and sized by salary - so you can spot both departmental clustering and which points represent higher earners at a glance.

```python
# Small dataset: employee age, salary, and department
df2 = pd.DataFrame({'age': [22, 25, 47, 52, 46], 'salary': [25000, 32000, 65000, 72000, 68000], 'dept': ['IT', 'IT', 'HR', 'HR', 'Sales']})
# color encodes department as a category, size scales each point by its salary value
fig = px.scatter(df2, x='age', y='salary', color='dept', size='salary', title='Age vs Salary')
fig.show()
```

## Interactive Bar Chart
Plotly Express follows the same "grammar of graphics" approach as seaborn - you describe what data maps to which visual attribute (x, y, color, etc.) rather than manually drawing shapes - so if you already know `sns.barplot`, `px.bar` will feel familiar. The difference is purely in the output: seaborn/matplotlib produce a static PNG-like image, while plotly produces an interactive HTML/JavaScript figure you can hover over and zoom into, which is why plotly is popular for dashboards and web-embedded reports. The code below reuses the `df2` employee data and builds a bar chart comparing salary across departments - each bar's height shows that department's salary value, and hovering over a bar reveals the exact number.

```python
# Bar chart comparing salary across departments, same df2 data as above
fig = px.bar(df2, x='dept', y='salary', title='Salary by Department')
fig.show()
```
