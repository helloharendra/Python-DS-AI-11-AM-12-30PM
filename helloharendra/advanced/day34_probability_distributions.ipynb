## Binomial Distribution
The binomial distribution models situations with a fixed number of independent yes/no (success/failure) trials, each with the same probability of success - like flipping a coin 10 times, or asking 10 customers if they'll buy a product. It's used constantly in data science for A/B testing, quality control (defect rates), and any scenario where you're counting "how many out of N" successes occurred. The code below computes the probability mass function (PMF) for `n=10` trials with `p=0.5` probability of success each - the PMF gives the exact probability of getting each possible count of successes, from 0 to 10. Plotting it as a bar chart shows a symmetric, bell-ish shape centered at 5 (since with a fair 50% chance, getting around half the trials as successes is most likely), with the bar height at each x-value representing that exact outcome's probability.

```python
from scipy import stats
import matplotlib.pyplot as plt

# n = number of trials, p = probability of success on each trial
n, p = 10, 0.5
x = range(0, n + 1)
# PMF: probability of getting exactly x successes, for each possible x
pmf = stats.binom.pmf(x, n, p)
plt.bar(x, pmf)
plt.title('Binomial Distribution (n=10, p=0.5)')
plt.show()
```

## Poisson Distribution
The Poisson distribution models how many times a rare, independent event happens in a fixed window of time or space, given a known average rate - classic examples are the number of customer support tickets per hour, website errors per day, or calls arriving at a call center per minute. It's defined by a single parameter, `mu` (mu), which is the average (expected) number of events in that interval. The code below sets `mu=3` (an average of 3 events per interval) and computes the PMF for possible event counts from 0 to 10, then plots it as a bar chart. Unlike the binomial's symmetric shape, the Poisson distribution is skewed right (a longer tail toward higher values) when the mean is small, with the tallest bar near the mean value of 3 and probabilities shrinking as you move further from it.

```python
# mu = average rate of events per interval (here, 3 events on average)
mu = 3
x = range(0, 11)
# PMF: probability of observing exactly x events, for each possible x
pmf = stats.poisson.pmf(x, mu)
plt.bar(x, pmf, color='orange')
plt.title('Poisson Distribution (mu=3)')
plt.show()
```

## Normal Distribution
The normal (Gaussian) distribution is the most important distribution in statistics because so many real-world measurements - heights, test scores, measurement errors - naturally cluster around an average with symmetric spread, and it's also the distribution that sample means tend toward as sample size grows (the Central Limit Theorem, covered in Day 35). Unlike the binomial or Poisson, the normal distribution is continuous, so instead of a PMF (probabilities of exact counts) we use a probability density function (PDF), which describes relative likelihood at each point along a continuous range rather than an exact probability at that single point. The code below computes the PDF of the "standard normal" distribution (mean 0, standard deviation 1, set via `loc=0, scale=1`) across a range of x-values from -4 to 4, and plots it as a smooth curve. The result is the familiar symmetric bell curve, tallest at 0 and tapering off toward both tails.

```python
import numpy as np
# 200 evenly spaced x-values from -4 to 4, enough points for a smooth curve
x = np.linspace(-4, 4, 200)
# PDF of the standard normal: loc is the mean, scale is the standard deviation
pdf = stats.norm.pdf(x, loc=0, scale=1)
plt.plot(x, pdf)
plt.title('Standard Normal Distribution')
plt.show()
```

## Reading Probabilities from a Distribution
While the PMF/PDF tells you the probability (or density) of an exact value, real questions are usually phrased as "at most" or "at least" something - like "what fraction of customers spend under $70?" That's what the cumulative distribution function (CDF) answers: `cdf(x)` gives P(X <= x), the total probability accumulated from the lowest possible value up through `x`. Because it's cumulative, the CDF is always non-decreasing and ranges from 0 to 1. Below, `stats.norm.cdf(1)` answers "what's the probability a standard normal value is at most 1?" (about 84%, since 1 standard deviation above the mean captures the bulk of the distribution), and `stats.poisson.cdf(6, mu)` answers "what's the probability of 6 or fewer events, given an average rate of 3?" - useful for questions like "how likely is it we get 6 or fewer support tickets today?"

```python
# CDF gives the cumulative probability P(X <= x), i.e. "probability of this value or lower"
print('P(X <= 1) for standard normal:', stats.norm.cdf(1))
print('P(X <= 6) for Poisson(mu=3):', stats.poisson.cdf(6, mu))
```
