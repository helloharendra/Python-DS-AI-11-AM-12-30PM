# Day 33: Lab - Data Storytelling Mini-Project

## Project Goal
Data storytelling is the skill of choosing the right chart for each question you want to answer, rather than dumping every possible plot on a dataset - a good analyst picks a small number of charts that each make one clear point. This lab practices that by taking a single small revenue dataset and building three different charts, each one purpose-built to answer a different question: how did revenue change over time, how do regions compare in total, and how much does revenue vary within each region. We start by creating the raw dataset - monthly revenue figures tagged with a region - which every chart in this notebook will reuse.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Six months of revenue, each month tagged with the region it belongs to
df = pd.DataFrame({
    'month': ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun'],
    'revenue': [12000, 15000, 11000, 18000, 21000, 19000],
    'region': ['East', 'East', 'West', 'West', 'East', 'West']
})
```

## Insight 1: Overall Trend
The first question in almost any business analysis is "is this metric going up or down over time?" - and a line chart is the clearest way to answer that, because connecting points in time order makes the direction and rate of change immediately visible, in a way a table of numbers can't. Here we plot `revenue` against `month` in chronological order, with markers at each data point so individual month values are still readable, not just the overall shape. Reading the chart left to right shows revenue rising and dipping across the six months - any sharp jump or drop stands out visually as a peak or valley, which is exactly the kind of pattern you'd want to call out first in a report.

```python
# Line chart with markers: shows revenue's month-over-month trend
plt.plot(df['month'], df['revenue'], marker='o')
plt.title('Revenue Trend Over 6 Months')
plt.ylabel('Revenue')
plt.show()
```

## Insight 2: Regional Comparison
Once you know the overall trend, the next natural question is "which region is driving that trend?" - and comparing totals across a small number of categories is exactly what a bar chart is best at, since bar height makes differences in magnitude easy to compare at a glance. To get there, we first use `groupby('region')['revenue'].sum()` to collapse the six monthly rows down into one total per region (East and West), which is the aggregation step that has to happen before a category-level comparison makes sense. Plotting that aggregated Series directly as a bar chart, with a distinct color per bar, then shows which region brought in more total revenue over the period - the taller bar is the stronger-performing region.

```python
# Sum revenue within each region to get one total per region
region_totals = df.groupby('region')['revenue'].sum()
# Plot the aggregated totals as a bar chart, one distinct color per region
region_totals.plot(kind='bar', color=['#4C72B0', '#DD8452'])
plt.title('Total Revenue by Region')
plt.show()
```

## Insight 3: Distribution Check
Totals alone can hide important detail - a region could hit the same total revenue either through consistent months or through one huge month offsetting several weak ones, and a bar chart of totals can't tell those two situations apart. A boxplot fixes this by showing the spread of values within each region: the box covers the middle 50% of values (25th to 75th percentile), the line inside marks the median, and the "whiskers" extend to the typical range, with any points far outside that range plotted individually as outliers. Using seaborn's `boxplot` with `x='region'` and `y='revenue'` draws one box per region straight from the raw (non-aggregated) data. A taller box or wider whiskers means that region's monthly revenue is more volatile, while a compact box means its months were more consistent.

```python
# Boxplot of raw (non-aggregated) revenue per region: shows spread, not just totals
sns.boxplot(x='region', y='revenue', data=df)
plt.title('Revenue Spread by Region')
plt.show()
```
