# Day 35: Sampling & Central Limit Theorem

## Population vs Sample
In statistics, the "population" is every possible unit you'd ideally want to measure (every customer, every transaction, every user), while a "sample" is the smaller, practical subset you actually collect data from - you almost never get to measure the whole population, so nearly all real analysis works with samples and tries to draw conclusions about the population from them. Here we simulate a population of 100,000 values so we have "ground truth" to compare our sample-based estimates against later in the notebook. We deliberately draw it from an exponential distribution (`np.random.exponential`), which is skewed (most values are small, with a long tail of rare large values, like customer wait times or purchase amounts) rather than symmetric like a normal distribution - this matters for the Central Limit Theorem demonstration coming up. `np.random.seed(0)` fixes the random number generator so the same "random" values are produced every time this cell runs, making the results reproducible. The printed mean is the true, exact average of the entire simulated population.

```python
import numpy as np

# Fix the random seed so this notebook produces identical results every run
np.random.seed(0)
# Simulate a skewed population of 100,000 values (exponential distribution,
# scale=50 sets the average value) - this stands in for "all possible data"
population = np.random.exponential(scale=50, size=100000)
print('Population mean:', population.mean())
```

## Taking Repeated Samples
To see how sample means behave, we need to actually draw many samples and track each one's average, rather than just drawing a single sample once. The list comprehension below repeats this process 1000 times: each iteration draws a random sample of 30 values from the population (`np.random.choice(population, size=30)`) and immediately computes that sample's mean, and all 1000 of those individual sample means get collected into the `sample_means` list. In real life you'd never actually draw 1000 samples of your own data - this simulation exists purely to let us visualize, in the next cell, how the means of many different samples are distributed, which is the foundation for understanding why a single sample's mean is a trustworthy estimate of the population mean.

```python
# Draw 1000 random samples of size 30 from the population and record each sample's mean
sample_means = [np.random.choice(population, size=30).mean() for _ in range(1000)]
```

## Central Limit Theorem
The Central Limit Theorem (CLT) is one of the most important results in statistics: no matter what shape the original population's distribution has (skewed, uniform, bimodal, etc.), if you repeatedly draw samples of a reasonably large size and compute their means, the distribution of those sample means will look approximately normal (bell-shaped), and will center on the true population mean. This is exactly what we're about to visualize directly: the first histogram plots the raw `population` data, which we deliberately built to be skewed (a long tail toward larger values, since it came from an exponential distribution). The second histogram plots the 1000 `sample_means` computed in the previous cell - even though it was built from the same skewed population, you should see it forms a much more symmetric, bell-like shape clustered tightly around the true mean. That contrast between a skewed source and a normal-looking sampling distribution is the CLT in action.

```python
import matplotlib.pyplot as plt

# The raw population is skewed - a long right tail of large values
plt.hist(population, bins=50, alpha=0.5, label='Population (skewed)')
plt.legend()
plt.title('Original Population Distribution')
plt.show()

# The distribution of sample MEANS is approximately normal, per the CLT,
# even though it was built from a skewed population
plt.hist(sample_means, bins=50, color='green')
plt.title('Distribution of Sample Means (looks normal)')
plt.show()
```

## Why It Matters
The practical payoff of the CLT is that it lets us use normal-distribution-based statistical tools - confidence intervals, z-tests, and many hypothesis tests - even when the underlying data itself isn't normally distributed, as long as the sample size is large enough (a common rule of thumb is n >= 30). Without the CLT, we'd need to know the exact shape of the population's distribution before we could make any reliable statistical statements about it, which is rarely possible in practice. The check below confirms the CLT's key promise numerically: the average of all 1000 sample means (`np.mean(sample_means)`) should come out very close to the true population mean, even though each individual sample only saw 30 out of the 100,000 population values. This is why sampling works - a well-drawn sample's mean is, on average, a good, unbiased estimate of the population mean.

```python
# The average of many sample means should land very close to the true population mean
print('Mean of sample means:', np.mean(sample_means))
print('True population mean:', population.mean())
```
