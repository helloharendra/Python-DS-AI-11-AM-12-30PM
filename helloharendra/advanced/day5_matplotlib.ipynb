# Day 5: Data Visualization with Matplotlib

## Line Plot
Line plots are the go-to chart for showing how a value changes across a continuous or ordered variable, most often time (stock prices, sensor readings, monthly revenue) — the connecting line makes trends and turning points easy to spot at a glance. `plt.plot(x, y, marker='o')` draws a line connecting each `(x, y)` pair in order and puts a circular marker at each actual data point so you can distinguish real measurements from the interpolated line between them; `title()`, `xlabel()`, and `ylabel()` add text labels, and `plt.show()` renders the figure. Looking at the output, you should see a line rising and falling as it connects the 5 points, with the y-values 10, 20, 15, 30, 25 tracing the shape of the trend.

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 30, 25]
plt.plot(x, y, marker='o')  # marker='o' draws a dot at each actual (x, y) data point
plt.title('Line Plot')
plt.xlabel('X')
plt.ylabel('Y')
plt.show()
```

## Bar Plot
Bar plots are used to compare a numeric value across distinct, unordered categories — like comparing sales across products, or counts across survey responses — where a line plot wouldn't make sense because there's no continuous progression between categories. `plt.bar(categories, values, color=...)` draws one rectangular bar per category, with the bar's height representing its value, making it easy to visually compare magnitudes side by side. In the output, you should see three bars labeled A, B, C, with bar B clearly the tallest (45) and bar C the shortest (25), matching the `values` list.

```python
categories = ['A', 'B', 'C']
values = [30, 45, 25]
plt.bar(categories, values, color='skyblue')  # one bar per category, height = value
plt.title('Bar Plot')
plt.show()
```

## Scatter Plot
Scatter plots reveal whether two numeric variables are related — for example, whether higher advertising spend correlates with higher sales — by plotting each pair of values as a single dot without connecting them, so you can see the overall pattern or spread rather than a forced sequence. `plt.scatter(x, y, color=...)` places one dot per `(x, y)` pair at its exact coordinate position. In the output, you should see 5 green dots that trend upward and to the right, suggesting that as `x` increases, `y` tends to increase too (a positive relationship).

```python
x = [5, 7, 8, 6, 9]
y = [10, 14, 13, 11, 18]
plt.scatter(x, y, color='green')  # one dot per (x, y) pair, no connecting line
plt.title('Scatter Plot')
plt.show()
```

## Histogram
A histogram shows how a single numeric variable is distributed — where most values cluster, whether the data is symmetric or skewed, and whether there are multiple peaks — which is one of the first things you check when exploring a new numeric column. `plt.hist(data, bins=20, color=...)` divides the range of `data` into 20 equal-width buckets ("bins") and draws a bar for each bin whose height is the count of values falling into it. Since `data` is generated with `np.random.randn` (standard normal distribution), the output should show the classic bell-curve shape: most values clustered around 0, tapering off symmetrically toward the edges.

```python
import numpy as np
data = np.random.randn(1000)  # 1000 values drawn from a standard normal distribution
plt.hist(data, bins=20, color='orange')  # bins=20 splits the range into 20 buckets to count values in
plt.title('Histogram')
plt.show()
```

## Subplots
Subplots let you place multiple related charts in a single figure, which is useful when you want to compare different views of the data side by side (like a line trend and a category comparison) without generating separate images. `plt.subplots(1, 2, figsize=(10, 4))` creates a figure with 1 row and 2 columns of chart areas ("axes") and returns them as an array (`axes`) you can plot into individually, using each axis's own `.plot()`/`.bar()`/`.set_title()` methods instead of the global `plt` functions. `plt.tight_layout()` automatically adjusts spacing so titles and labels don't overlap; the output shows a line chart on the left and a bar chart on the right, sharing one figure.

```python
# Create a figure with 1 row, 2 columns of subplots (axes[0] = left chart, axes[1] = right chart)
fig, axes = plt.subplots(1, 2, figsize=(10, 4))
axes[0].plot([1, 2, 3], [1, 4, 9])
axes[0].set_title('Line')
axes[1].bar(['A', 'B'], [10, 20])
axes[1].set_title('Bar')
plt.tight_layout()  # auto-adjust spacing so titles/labels don't overlap
plt.show()
```
