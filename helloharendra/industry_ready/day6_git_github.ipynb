# Day 6: Git & GitHub for Data Teams

## Git Basics
git tracks changes to your project locally, in a hidden .git folder, one commit (snapshot) at a time.

```python
# terminal commands:
# git init                       # start tracking a folder
# git status                     # see what changed
# git add file.py                # stage a specific file
# git add .                      # stage everything changed
# git commit -m "Add data cleaning script"
# git log --oneline              # see commit history
```

## Branching & Merging
Branches let you work on a feature or experiment without touching the main/working code.

```python
# terminal commands:
# git branch feature/xgboost-model     # create a branch
# git checkout feature/xgboost-model   # switch to it
# git checkout -b feature/eda-report   # create + switch in one step
# ...make changes, commit...
# git checkout main
# git merge feature/xgboost-model      # bring the changes into main
# git branch -d feature/xgboost-model  # delete the merged branch
```

## Remote Repositories (GitHub)
The remote is the copy hosted on GitHub that your team shares - clone it, and push/pull to sync.

```python
# terminal commands:
# git clone https://github.com/username/repo.git
# git remote add origin https://github.com/username/repo.git
# git push -u origin main       # first push, sets upstream
# git push                      # later pushes
# git pull                      # fetch + merge teammates' latest changes
```

## .gitignore for Data Science Projects
Never commit datasets, secrets, or environment folders - a .gitignore keeps them out automatically.

```python
# .gitignore contents for a typical DS/ML repo:
# venv/
# __pycache__/
# *.pyc
# .ipynb_checkpoints/
# *.csv
# *.xlsx
# *.pkl
# .env
# .DS_Store
# mlruns/
```

## Clean Notebook Diffs & Pull Requests
Committed notebook outputs (plots, big JSON blobs) make diffs unreadable - clear outputs before committing, and use PRs for review.
