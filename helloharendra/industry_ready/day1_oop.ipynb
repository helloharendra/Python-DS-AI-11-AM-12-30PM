# Day 1: Object-Oriented Programming in Python

## Classes & Objects
A class is a blueprint; an object is an instance of it. __init__ runs when an object is created.

```python
class BankAccount:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
    
        self.balance += amount
        return self.balance

    def withdraw(self, amount):
        if amount > self.balance:
            raise ValueError("Insufficient funds")
        self.balance -= amount
        return self.balance

acc = BankAccount("Harendra", 0)
acc.deposit(2000)
acc.deposit(4000)
acc.deposit(1000)
acc.withdraw(2000)
acc.withdraw(1500)

print(acc.owner, acc.balance)
```

## Encapsulation
Prefix an attribute with _ (convention) or __ (name-mangled) to signal it's internal, then expose it safely through a @property.

```python
class BankAccount:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.__balance = balance   # name-mangled, not directly accessible as acc.__balance

    @property
    def balance(self):
        return self.__balance

    @balance.setter
    def balance(self, value):
        if value < 0:
            raise ValueError("Balance cannot be negative")
        self.__balance = value

acc = BankAccount("Harendra", 1000)
print(acc.balance)
acc.balance = 2000
print(acc.balance)
```

## Inheritance
A child class reuses and extends a parent class using super() to call the parent's constructor/methods.

```python
class SavingsAccount(BankAccount):
    def __init__(self, owner, balance=0, rate=0.04):
        super().__init__(owner, balance)
        self.rate = rate

    def add_interest(self):
        interest = self.balance * self.rate
        self.balance += interest
        return interest

sav = SavingsAccount("Harendra", 1000, rate=0.05)
print(sav.add_interest())
print(sav.balance)
```

## Polymorphism & Method Overriding
Different classes can implement the same method name differently; calling code doesn't need to know which subclass it has.

```python
class CurrentAccount(BankAccount):
    def account_type(self):
        return "Current Account (no interest)"

class SavingsAccount2(BankAccount):
    def account_type(self):
        return "Savings Account (earns interest)"

for acc in [CurrentAccount("A", 100), SavingsAccount2("B", 200)]:
    print(acc.owner, "->", acc.account_type())
```

## Dunder (Magic) Methods
Implementing __str__, __repr__, __eq__ and __len__ lets your objects work naturally with print(), ==, and len().

```python
class Point:
    def __init__(self, x, y):
        self.x, self.y = x, y

    def __str__(self):
        return f"({self.x}, {self.y})"

    def __repr__(self):
        return f"Point(x={self.x}, y={self.y})"

    def __eq__(self, other):
        return self.x == other.x and self.y == other.y

p1, p2 = Point(1, 2), Point(1, 2)
print(p1)          # uses __str__
print(repr(p1))    # uses __repr__
print(p1 == p2)    # uses __eq__
```

## Class Methods & Static Methods
@classmethod receives the class itself (useful as alternate constructors); @staticmethod is a plain function namespaced inside a class.

```python
class Employee:
    def __init__(self, name, salary):
        self.name, self.salary = name, salary

    @classmethod
    def from_csv_row(cls, row):           # alternate constructor
        name, salary = row.split(",")
        return cls(name, float(salary))

    @staticmethod
    def annual_ctc(monthly_salary):       # utility, doesn't need self/cls
        return monthly_salary * 12

e = Employee.from_csv_row("Riya,45000")
print(e.name, e.salary)
print(Employee.annual_ctc(e.salary))
```
