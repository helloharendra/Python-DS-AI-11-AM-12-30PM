## Why Modules?
Splitting code into files (modules) keeps projects organized and reusable instead of one giant script.

```python
# math_utils.py  (save this as a separate file in your project)
def is_even(n):
    return n % 2 == 0

def average(numbers):
    return sum(numbers) / len(numbers)
```

## Importing Your Own Module
Once math_utils.py sits next to your notebook/script, import it like any built-in module.

```python
# in another file, e.g. main.py, in the same folder as math_utils.py
# import math_utils
# print(math_utils.is_even(10))
# print(math_utils.average([10, 20, 30]))

# import styles worth knowing:
# import math_utils                  -> math_utils.is_even(10)
# import math_utils as mu            -> mu.is_even(10)
# from math_utils import is_even     -> is_even(10)
# from math_utils import *           -> avoid: pollutes the namespace
```

## Package Structure
A package is a folder of modules with an __init__.py, letting you group related modules under one importable name.

```python
# project layout for a small package:
# my_dsa_toolkit/
#     __init__.py
#     sorting.py
#     searching.py
#
# __init__.py can be empty, or re-export the pieces you want public:
#     from .sorting import merge_sort
#     from .searching import binary_search
#
# then anywhere in the project:
# from my_dsa_toolkit import merge_sort
```

## pip & requirements.txt
pip installs third-party packages; requirements.txt freezes exact versions so a teammate (or production server) gets the same environment.

```python
# terminal commands (not Python):
# pip install pandas scikit-learn fastapi
# pip freeze > requirements.txt          # snapshot current environment
# pip install -r requirements.txt        # recreate it elsewhere

# example requirements.txt content:
# pandas==2.2.2
# scikit-learn==1.5.0
# fastapi==0.111.0
# uvicorn==0.30.1
```

## Virtual Environments
A virtual environment isolates a project's dependencies so different projects can use different (even conflicting) package versions.

```python
# terminal commands (not Python):
# python -m venv venv                 # create an environment named "venv"
# source venv/bin/activate            # activate it (Mac/Linux)
# venv\Scripts\activate               # activate it (Windows)
# pip install -r requirements.txt
# deactivate                          # leave the environment

# conda alternative:
# conda create -n ds-course python=3.11
# conda activate ds-course
```

## if __name__ == "__main__"
This guard lets a file be both a reusable, importable module AND a runnable script, without the script part firing on import.

```python
# script.py
def main():
    print("Running as the main program")

if __name__ == "__main__":
    main()

# If another file does `import script`, main() will NOT run automatically.
# If you run `python script.py` directly, it WILL run.
```
