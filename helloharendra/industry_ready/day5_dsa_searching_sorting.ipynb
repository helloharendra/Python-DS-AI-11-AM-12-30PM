# Day 5: DSA — Searching & Sorting Algorithms

## Linear Search — O(n)
Checks every element one by one. Works on unsorted data but scales poorly.

```python
def linear_search(arr, target):
    for i, value in enumerate(arr):
        if value == target:
            return i
    return -1

print(linear_search([5, 3, 8, 1, 9], 8))
```

## Binary Search — O(log n)
Repeatedly halves the search space - but the array must be sorted first.

```python
def binary_search(sorted_arr, target):
    low, high = 0, len(sorted_arr) - 1
    while low <= high:
        mid = (low + high) // 2
        if sorted_arr[mid] == target:
            return mid
        elif sorted_arr[mid] < target:
            low = mid + 1
        else:
            high = mid - 1
    return -1

data = [1, 3, 5, 7, 9, 11, 13]
print(binary_search(data, 9))
```

## Bubble Sort — O(n²)
Repeatedly swaps adjacent out-of-order pairs. Simple to understand, rarely used in production.

```python
def bubble_sort(arr):
    arr = arr.copy()
    n = len(arr)
    for i in range(n):
        for j in range(0, n - i - 1):
            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
    return arr

print(bubble_sort([5, 1, 4, 2, 8]))
```

## Merge Sort — O(n log n)
Divide-and-conquer: split the array in half, sort each half, then merge the sorted halves.

```python
def merge_sort(arr):
    if len(arr) <= 1:
        return arr
    mid = len(arr) // 2
    left, right = merge_sort(arr[:mid]), merge_sort(arr[mid:])

    merged, i, j = [], 0, 0
    while i < len(left) and j < len(right):
        if left[i] <= right[j]:
            merged.append(left[i]); i += 1
        else:
            merged.append(right[j]); j += 1
    merged.extend(left[i:])
    merged.extend(right[j:])
    return merged

print(merge_sort([5, 1, 4, 2, 8, 9, 3]))
```

## Quick Sort — O(n log n) average
Picks a pivot, partitions everything smaller/larger around it, then sorts each side recursively.

```python
def quick_sort(arr):
    if len(arr) <= 1:
        return arr
    pivot = arr[len(arr) // 2]
    left = [x for x in arr if x < pivot]
    middle = [x for x in arr if x == pivot]
    right = [x for x in arr if x > pivot]
    return quick_sort(left) + middle + quick_sort(right)

print(quick_sort([5, 1, 4, 2, 8, 9, 3]))
```

## Choosing the Right Algorithm
In practice you rarely hand-roll these — Python's built-in sorted()/list.sort() use Timsort (O(n log n)) — but interviews and CS fundamentals expect you to know how and why.

```python
# Built-in sort (Timsort) - what you'd actually use day to day:
data = [5, 1, 4, 2, 8, 9, 3]
print(sorted(data))                       # ascending
print(sorted(data, reverse=True))         # descending
print(sorted(data, key=lambda x: -x))     # custom key

# Rule of thumb:
# Linear search  -> unsorted, small data
# Binary search  -> sorted, large data, frequent lookups
# Bubble sort    -> teaching only
# Merge sort     -> stable, guaranteed O(n log n), needs extra memory
# Quick sort     -> fast in practice, in-place, worst case O(n^2)
```
