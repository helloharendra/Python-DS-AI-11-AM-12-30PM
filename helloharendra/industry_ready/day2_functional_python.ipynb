## Lambda Functions
A lambda is a small, unnamed function for a single expression - handy for a quick sort key or a one-off transform.

```python
square = lambda x: x ** 2
print(square(5))

students = [("Amit", 72), ("Riya", 91), ("Zoya", 65)]
students_sorted = sorted(students, key=lambda s: s[1], reverse=True)
print(students_sorted)
```

## map(), filter(), reduce()
map() transforms every item, filter() keeps items matching a condition, reduce() folds a sequence into one value.

```python
from functools import reduce

marks = [45, 78, 92, 33, 61]
percentages = list(map(lambda m: m / 100, marks))
passed = list(filter(lambda m: m >= 40, marks))
total = reduce(lambda a, b: a + b, marks)

print(percentages)
print(passed)
print(total)
```

## *args and **kwargs
*args collects extra positional arguments into a tuple; **kwargs collects extra keyword arguments into a dict.

```python
def summarize(*scores, **student_info):
    print("Student info:", student_info)
    print("Average score:", sum(scores) / len(scores))

summarize(80, 90, 70, name="Aisha", grade="10th")
```

## Closures
An inner function that remembers variables from its enclosing scope even after that scope has finished running.

```python
def make_multiplier(factor):
    def multiplier(x):
        return x * factor
    return multiplier

double = make_multiplier(2)
triple = make_multiplier(3)
print(double(10), triple(10))
```

## Decorators
A decorator wraps a function to add behaviour (logging, timing, access checks) without changing the function's own code.

```python
import functools

def log_call(func):
    @functools.wraps(func)
    def wrapper(*args, **kwargs):
        print(f"Calling {func.__name__} with args={args} kwargs={kwargs}")
        result = func(*args, **kwargs)
        print(f"{func.__name__} returned {result}")
        return result
    return wrapper

@log_call
def add(a, b):
    return a + b

add(3, 4)
```

## Practical Decorator — Timing a Data Pipeline Step
A reusable @timeit decorator is a common real-world pattern for profiling slow ETL/feature-engineering steps.

```python
import time, functools

def timeit(func):
    @functools.wraps(func)
    def wrapper(*args, **kwargs):
        start = time.perf_counter()
        result = func(*args, **kwargs)
        elapsed = time.perf_counter() - start
        print(f"{func.__name__} took {elapsed:.4f}s")
        return result
    return wrapper

@timeit
def load_and_square(n):
    return [i ** 2 for i in range(n)]

load_and_square(1_0)
```
