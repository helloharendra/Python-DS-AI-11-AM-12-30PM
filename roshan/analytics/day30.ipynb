```python
import pandas as pd
import numpy as np 
import matplotlib.pyplot as pl
```

```python
store = pd.DataFrame({
    'cust_id':[101,102,103,104,105,106],
    'cust_name':['Rahul','Rohit','Ram','Satish','Amit','Prashant'],
    'city':['Azamgarh','Sitapur','Azamgarh','Varanasi','Varanasi','Ambedkar nagar'],
    'product':['samsung','apple','apple','motorola','realme','oppo'],
    'price':[25000,35000,22000,40000,18000,23000],
    'quantity':[10,12,15,5,20,14]
})
```

```python
store
```

```python
store['total_amount'] = store['price']*store['quantity']
print(store)
```

```python
grouped = store.groupby('city')['total_amount'].mean()
```

```python
grouped
```

```python
grouped_max = store.groupby('city')['quantity'].max()
print(grouped_max)
```

```python
agg = store.groupby('product').agg({
    'price':['min','max','mean'],
    'total_amount':['mean']
})
print(agg)
```

```python
size = store.groupby('city').size()
print(size)
```

```python
customer = pd.DataFrame({
    'id' :[101,102,103],
    'name' :['roshan','vikash','prashant'],
    'marks' :[45,55,65] 
})
customers1 =pd.DataFrame({
    'id' :[101,102,104],
    'age' :[18,20,25],
    'class':[10,12,11]
})

```

```python
merge = pd.merge(customer,customers1, on='id',how='inner')
```

```python
merge
```

```python
merge_left = pd.merge(customer,customers1, on='id',how='left')
print(merge_left)
```

```python
merge_right = pd.merge(customer,customers1, on='id',how='right')
print(merge_right)
```

```python
merge_outer = pd.merge(customer,customers1, on='id',how='outer')
print(merge_outer)
```

```python
concatnate_column = pd.concat([customer,customers1],axis=0)
print(concatnate_column)
```

```python
concatnate_rows = pd.concat([customer,customers1],axis=1)
print(concatnate_rows)
```
