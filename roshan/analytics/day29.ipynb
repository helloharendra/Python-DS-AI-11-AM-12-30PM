```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
employee_data = pd.DataFrame({
    'Name':['Roshan','Vikash','Amit','Prashant'],
    'Age':[25,20,22,21],
    'Score':[45,65,75,85]
})
print(employee_data)
```

```python
bar = employee_data.plot(
    kind='bar',
    x='Name',
    y='Score',
    title='Score by employee'
)
```

```python
employee_data1 = pd.DataFrame({
    'Name':['Roshan','Vikash','Amit','Prashant'],
    'Age':[25,20,22,21],
    'Score':[45,65,75,85]
})
bar = employee_data1.plot(
    kind='bar',
    x='Name',
    y='Score',
    title='Score by employee'
)
employee_data1.plot(
    kind='line',
    x='Name',
    y='Score',
    title='Score by employee'
)


employee_data1.plot(
    kind='pie',
    x='Name',
    y='Score',
    labels=employee_data1['Name'],
    autopct="%1.1f%%",
    title='Score by employee',
   
)
```

```python
store = pd.DataFrame({
    'cust_id':[101,102,103,104,105,106],
    'cust_name':['Rahul','Rohit','Ram','Satish','Amit','Prashant'],
    'product':['samsung','nokia','apple','motorola','realme','oppo'],
    'price':[25000,35000,22000,40000,18000,23000],
    'quantity':[10,12,15,5,20,14]
})
```

```python
store['Total_amount']=store['price']*store['quantity']
```

```python

store.plot(
    kind='bar',
    x='cust_name',
    y='Total_amount',
    title='total amount by customer'
)
store.plot(
    kind='line',
    x='product',
    y='Total_amount',
    title='total amount by customer'
)
store.plot(
    kind='pie',
    x='product',
    y='quantity',
    autopct="%1.1f%%",
    labels=store['product'],
    title='Total amount by product'
)
store.plot(
    kind='barh',
    x='product',
    y='quantity',
  
    title='Total amount by product'
)
store.plot(
    kind='density',
    x='product',
    y='quantity',
  
    title='Total amount by product'
)
store.plot(
    kind='area',
    x='product',
    y='quantity',
  
    title='Total amount by product'
)
```
