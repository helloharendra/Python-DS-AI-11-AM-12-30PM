```python
import cv2
cap = cv2.VideoCapture(0)

while True:
    ret, frame = cap.read()

    if not ret:
        break 
    cv2.flip(frame,1)
    cv2.imshow("Mirror Face Detection",frame)


    if cv2.waitKey(1) & 0xFF==ord('q'):
        break
cap.release()
cv2.destroyAllWindows()
```

```python
def factorial(n):
    if n==1:
        return 1
    else:
        return n*factorial(n-1)
```

```python
print(factorial(5))
```
