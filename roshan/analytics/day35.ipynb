### mean median 

```python
marks = np.array([72,85,91,64,78,88,55,93,70,82])
print('marks :',marks)
print('marks :',np.mean(marks))
print('marks :',np.median(marks))
print('marks :',np.std(marks),2)
print('marks :',np.var(marks),2)
print('marks :',np.min(marks))
print('marks :',np.max(marks))
print('marks :',np.sum(marks))
print('marks :',np.percentile(marks),25)
print('marks :',np.percentile(marks),75)
print('marks range :',np.ptp(marks))
```

```python
sales = np.array([
    [100,200,150],
    [130,180,160],
    [140,220,170]
])
print(sales)
```
