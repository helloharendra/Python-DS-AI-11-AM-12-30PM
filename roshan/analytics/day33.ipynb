##Slicing and Indexing

```python
marks = np.array([75,65,45,95,95,75,65,25,35,45])
```

```python
print('all marks:',marks)
print('first marks:',marks[0])
print('last marks:',marks[-1])
print('2 gaping marks traversed:',marks[::2])
print('slicicing- 2 to 5:',marks[1:4])
print('reversed marks:',marks[::-1])
```
