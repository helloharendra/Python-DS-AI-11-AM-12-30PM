```python
data = pd.DataFrame({
    'product':['Mouse','Keyboard','Mouse','Keyboard'],
    'sales':[250,350,450,620],
    'region':['East','West','East','West']
})
```

```python
pivot = pd.pivot_table(data ,
        values='sales',
        index='product',
        aggfunc='sum',
        columns='region',
        fill_value=0,
        margins=True

)
```

```python
def grade(score):
    if score>=90:
        return 'A'
    if score>=80:
        return 'B'
    else:
        return 'C'
```

```python
customer['grade']=customer['score'].apply(grade)
print(customer[['name','score','grade']])
```
