```python
sale = {
    'month':['jan','feb','march','apr'],
    'sales':[2000,2500,1500,1700]
}
```

```python
df = pd.DataFrame(sale)
print(df)
```

```python
plot = df.plot(
    kind='bar',
    x='month',
    y='sales',
    figsize=(8,5)
)
plt.show()
```

```python
plot1 = df.plot(
    kind='line',
    marker="o",
    x='month',
    y='sales',
    figsize=(8,5)  
)
```

```python
plot2 = df.set_index('month')['sales'].plot(
    kind='pie',
    x='month',
    autopct='%1.1f%%',
    title='sales month wise',
    figsize=(6,6)  
)
```

```python
exercise = pd.DataFrame({
   'Days':["Mon","Tue","Wed","Thu","Fri"],
    'Visitors':[100,120,150,180,200]
})
```

```python
bar = exercise.plot(
    kind='bar',
    x='Days',
    y='Visitors',
    label='Visitor by Days'
)
```

```python
line = exercise.plot(
    kind='line',
    x='Days',
    y='Visitors',
    label='Visitor by Days'
)
```

```python
pie = exercise.set_index('Days')['Visitors'].plot(
    kind='pie',
    x='Days',
    
    autopct='%1.1F%%',
    label='Visitor by Days'
)
```
